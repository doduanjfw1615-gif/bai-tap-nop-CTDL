## Bài tập (tr. 31) — Vẽ BST và liệt kê theo 4 thứ tự

Cho dãy số, nhập lần lượt từ trái sang phải:

$$15;\ 30;\ 46;\ 3;\ 21;\ 7;\ 9;\ 10;\ 29;\ 34;\ 13;\ 5;\ 37;\ 18$$

**Yêu cầu:**
1. Vẽ cây nhị phân tìm kiếm (BST)
2. Liệt kê theo **PreOrder** (NLR)
3. Liệt kê theo **InOrder** (LNR)
4. Liệt kê theo **PostOrder** (LRN)
5. Liệt kê theo **Breath-first** (BFS — duyệt theo mức)

### Quy tắc tạo cây
Xuất phát từ gốc: nếu node cần thêm **nhỏ hơn** node đang xét thì thêm về **bên trái**, ngược lại thêm về **bên phải**.

### Cây kết quả (vẽ bằng text)

```
                    15
                  /    \
                 3      30
                  \    /  \
                   7  21   46
                  /      \   /
                 5        29 34
                            /  \
                           18   37
                  (và 9, 10, 13 nằm trên nhánh trái của 7)
```

Cây đầy đủ được in tự động ở cell bên dưới.

````python
class Node:
    """Một nút của cây nhị phân (theo đúng khai báo trong slide tr.19)."""
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


class BST:
    """Cây nhị phân tìm kiếm."""

    def __init__(self):
        self.root = None

    # ---------- 1. Khởi tạo / kiểm tra rỗng ----------
    def is_empty(self):
        return self.root is None

    # ---------- 3. Tạo nút & 4. Thêm vào cây ----------
    def insert(self, key):
        """Chèn key vào BST; bỏ qua nếu key đã tồn tại. Trả về True nếu chèn được."""
        if self.root is None:
            self.root = Node(key)
            return True
        return self._insert_rec(self.root, key)

    def _insert_rec(self, node, key):
        if key < node.key:
            if node.left is None:
                node.left = Node(key)
                return True
            return self._insert_rec(node.left, key)
        if key > node.key:
            if node.right is None:
                node.right = Node(key)
                return True
            return self._insert_rec(node.right, key)
        return False                     # trùng giá trị -> không chèn

    # ---------- 7. Duyệt cây ----------
    def preorder(self):
        out = []
        self._preorder(self.root, out)
        return out

    def _preorder(self, node, out):
        if node is not None:
            out.append(node.key)         # N
            self._preorder(node.left, out)   # L
            self._preorder(node.right, out)  # R

    def inorder(self):
        out = []
        self._inorder(self.root, out)
        return out

    def _inorder(self, node, out):
        if node is not None:
            self._inorder(node.left, out)    # L
            out.append(node.key)             # N
            self._inorder(node.right, out)   # R

    def postorder(self):
        out = []
        self._postorder(self.root, out)
        return out

    def _postorder(self, node, out):
        if node is not None:
            self._postorder(node.left, out)  # L
            self._postorder(node.right, out) # R
            out.append(node.key)             # N

    def breadth_first(self):
        """Duyệt theo chiều rộng (từng mức), dùng hàng đợi."""
        if self.root is None:
            return []
        out, hang_doi = [], [self.root]
        while hang_doi:
            node = hang_doi.pop(0)
            out.append(node.key)
            if node.left:
                hang_doi.append(node.left)
            if node.right:
                hang_doi.append(node.right)
        return out

    # ---------- 8. Tìm kiếm ----------
    def search(self, key):
        node = self.root
        while node is not None:
            if key == node.key:
                return node
            node = node.left if key < node.key else node.right
        return None

    # ---------- 9. Xóa cây ----------
    def clear(self):
        self.root = None

    # ---------- Hỗ trợ vẽ ----------
    def levels(self):
        """Trả về danh sách các mức, mỗi mức là list giá trị (None = khuyết)."""
        if self.root is None:
            return []
        res, cur = [], [self.root]
        while any(n is not None for n in cur):
            res.append([n.key if n else None for n in cur])
            nxt = []
            for n in cur:
                nxt.append(n.left if n else None)
                nxt.append(n.right if n else None)
            cur = nxt
        return res

    def draw(self):
        """In cây dạng text."""
        lv = self.levels()
        if not lv:
            print("(cây rỗng)")
            return
        width = 2 ** len(lv)
        for i, row in enumerate(lv):
            cell = 2 ** (len(lv) - i)
            line = ""
            for v in row:
                line += (str(v).center(cell) if v is not None else " " * cell)
            print(line.rstrip())
        print()


# ---- Dựng cây theo đúng dãy số của bài tập ----
day_so = [15, 30, 46, 3, 21, 7, 9, 10, 29, 34, 13, 5, 37, 18]

tree = BST()
for x in day_so:
    tree.insert(x)

print("Dãy nhập:", day_so)
print(f"Đã chèn {len(day_so)} giá trị (không có giá trị trùng).\n")
print("Cây BST:")
tree.draw()
````

````python
# ---- KẾT QUẢ 4 CÁCH DUYỆT ----
print("1. PreOrder  (NLR):", tree.preorder())
print("2. InOrder   (LNR):", tree.inorder())
print("3. PostOrder (LRN):", tree.postorder())
print("4. Breadth-first  :", tree.breadth_first())

print("\nNhận xét: InOrder của BST luôn cho dãy TĂNG DẦN.")
print("Dãy đã sắp xếp:", sorted(day_so))
````

````python
# ---- KIỂM THỬ ----
# 1) InOrder của BST phải bằng dãy đã sắp xếp
assert tree.inorder() == sorted(day_so), "InOrder phải là dãy tăng dần"
print("Test InOrder == sorted(day_so): PASS")

# 2) Mọi giá trị trong 4 cách duyệt đều là hoán vị của dãy nhập
for ten, kq in [("PreOrder", tree.preorder()), ("InOrder", tree.inorder()),
                ("PostOrder", tree.postorder()), ("BFS", tree.breadth_first())]:
    assert sorted(kq) == sorted(day_so), f"{ten} không phải hoán vị của dãy nhập"
print("Test cả 4 cách duyệt đều là hoán vị của dãy nhập: PASS")

# 3) BFS phải duyệt theo mức: mức trước xuất hiện trước mức sau
lv = tree.levels()
bfs_muc = [v for row in lv for v in row if v is not None]
assert tree.breadth_first() == bfs_muc, "BFS phải duyệt theo từng mức"
print("Test BFS == duyệt theo mức: PASS")

# 4) Tính chất BST: với mọi nút, giá trị trái < nút < giá trị phải
def kiem_tra_bst(node, low=float('-inf'), high=float('inf')):
    if node is None:
        return True
    if not (low < node.key < high):
        return False
    return (kiem_tra_bst(node.left, low, node.key) and
            kiem_tra_bst(node.right, node.key, high))

assert kiem_tra_bst(tree.root), "Vi phạm tính chất BST"
print("Test tính chất BST (low < key < high) trên mọi nút: PASS")

# 5) Duyệt lại bằng vòng lặp (stack) để đối chiếu độc lập với bản đệ quy
def preorder_stack(root):
    if root is None:
        return []
    out, st = [], [root]
    while st:
        n = st.pop()
        out.append(n.key)
        if n.right: st.append(n.right)
        if n.left:  st.append(n.left)
    return out

assert preorder_stack(tree.root) == tree.preorder()
print("Test PreOrder đệ quy == PreOrder dùng stack: PASS")

# 6) Tìm kiếm
assert all(tree.search(x) is not None for x in day_so)
assert tree.search(999) is None
print("Test tìm kiếm (có/không có): PASS")
````

## Mục luyện tập — Các tác vụ trên cây nhị phân (slide tr. 18)

Slide liệt kê 9 tác vụ: khởi tạo, kiểm tra rỗng, tạo nút, thêm trái/phải, **xóa trái/phải**, duyệt, tìm kiếm, xóa cây. Phần trên đã có hầu hết; dưới đây bổ sung **thêm/xóa cây con trái-phải**, **thông tin cây** (chiều cao, số nút, số lá, bậc) và **xóa cây**.

````python
class BST2(BST):
    """Bổ sung: thêm/xóa cây con trái-phải và các hàm thông tin của cây."""

    # ---- 4. Thêm trái / phải (theo đúng tác vụ trong slide) ----
    def add_left(self, parent_key, key):
        p = self.search(parent_key)
        if p is None or p.left is not None:
            return False
        p.left = Node(key)
        return True

    def add_right(self, parent_key, key):
        p = self.search(parent_key)
        if p is None or p.right is not None:
            return False
        p.right = Node(key)
        return True

    # ---- 6. Xóa trái / phải ----
    def remove_left(self, parent_key):
        p = self.search(parent_key)
        if p is None or p.left is None:
            return False
        p.left = None          # cả cây con bị bỏ
        return True

    def remove_right(self, parent_key):
        p = self.search(parent_key)
        if p is None or p.right is None:
            return False
        p.right = None
        return True

    # ---- 9. Xóa cây ----
    def clear(self):
        self.root = None

    # ---- Thông tin của cây ----
    def count_nodes(self, node=None, first=True):
        if first:
            node = self.root
        if node is None:
            return 0
        return 1 + self.count_nodes(node.left, False) + self.count_nodes(node.right, False)

    def height(self, node=None, first=True):
        """Chiều cao = số mức (nút gốc có chiều cao 1); cây rỗng = 0."""
        if first:
            node = self.root
        if node is None:
            return 0
        return 1 + max(self.height(node.left, False), self.height(node.right, False))

    def count_leaves(self, node=None, first=True):
        if first:
            node = self.root
        if node is None:
            return 0
        if node.left is None and node.right is None:
            return 1
        return self.count_leaves(node.left, False) + self.count_leaves(node.right, False)

    def count_degree(self):
        """Đếm số nút bậc 0 (lá), bậc 1, bậc 2."""
        b0 = b1 = b2 = 0
        def walk(n):
            nonlocal b0, b1, b2
            if n is None:
                return
            con = (n.left is not None) + (n.right is not None)
            if con == 0: b0 += 1
            elif con == 1: b1 += 1
            else: b2 += 1
            walk(n.left); walk(n.right)
        walk(self.root)
        return b0, b1, b2

    def find_min(self):
        if self.root is None:
            return None
        n = self.root
        while n.left is not None:
            n = n.left
        return n.key

    def find_max(self):
        if self.root is None:
            return None
        n = self.root
        while n.right is not None:
            n = n.right
        return n.key

    def nodes_at_level(self, k):
        """Các nút ở mức k (gốc là mức 0)."""
        out = []
        def walk(n, d):
            if n is None:
                return
            if d == k:
                out.append(n.key)
                return
            walk(n.left, d + 1); walk(n.right, d + 1)
        walk(self.root, 0)
        return out


t2 = BST2()
for x in day_so:
    t2.insert(x)

print("Số nút          :", t2.count_nodes())
print("Chiều cao       :", t2.height())
print("Số nút lá       :", t2.count_leaves())
print("Số nút bậc 0/1/2:", t2.count_degree())
print("Min / Max       :", t2.find_min(), "/", t2.find_max())
print("Các nút theo mức:")
for k in range(t2.height()):
    print(f"   mức {k}: {t2.nodes_at_level(k)}")
````

````python
# ---- KIỂM THỬ ----
# 1) Số nút, chiều cao, số lá đối chiếu với cách tính độc lập từ danh sách mức
lv = t2.levels()
assert t2.count_nodes() == sum(1 for row in lv for v in row if v is not None)
assert t2.height() == len(lv)

# đếm lá độc lập: nút không có con nào
def la_doc_lap(node):
    if node is None: return []
    if node.left is None and node.right is None: return [node.key]
    return la_doc_lap(node.left) + la_doc_lap(node.right)
assert t2.count_leaves() == len(la_doc_lap(t2.root))
print("Test số nút / chiều cao / số lá: PASS")

# 2) Định lý cây nhị phân: số lá = số nút bậc 2 + 1  (n0 = n2 + 1)
b0, b1, b2 = t2.count_degree()
assert b0 == b2 + 1, f"n0={b0} phải bằng n2+1={b2+1}"
print(f"Test định lý n0 = n2 + 1  ({b0} = {b2} + 1): PASS")

# 3) Min = phần tử nhỏ nhất, Max = phần tử lớn nhất
assert t2.find_min() == min(day_so) and t2.find_max() == max(day_so)
print("Test Min/Max == min/max dãy nhập: PASS")

# 4) Thêm/xóa cây con
t3 = BST2()
t3.insert(10)
assert t3.add_left(10, 5) and t3.add_right(10, 20)
assert not t3.add_left(10, 7)          # đã có con trái -> phải trả False
assert t3.preorder() == [10, 5, 20]
assert t3.remove_left(10) and t3.preorder() == [10, 20]
assert t3.remove_right(10) and t3.preorder() == [10]
print("Test add_left/add_right/remove_left/remove_right: PASS")

# 5) Xóa cây
t3.clear()
assert t3.is_empty() and t3.count_nodes() == 0 and t3.height() == 0
print("Test clear() -> cây rỗng: PASS")

# 6) Số nút ở tất cả các mức cộng lại = tổng số nút
assert sum(len(t2.nodes_at_level(k)) for k in range(t2.height())) == t2.count_nodes()
print("Test tổng nút các mức == tổng số nút: PASS")
````

## Mục luyện tập — Ứng dụng của cây (slide tr. 33–38)

Slide nêu 7 ứng dụng: BST, Trie, cây thư mục, Heap, cây quyết định, cây biểu thức, cây phân đoạn. Dưới đây cài đặt **3 ứng dụng kinh điển** có thể kiểm thử được: Trie (từ điển), Heap (hàng đợi ưu tiên), và cây biểu thức.

````python
# ---------- Ứng dụng 1: Trie (cây tiền tố) ----------
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_word = False


class Trie:
    """Cây tiền tố: chèn từ, tìm từ, tìm theo tiền tố (autocomplete)."""

    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        n = self.root
        for ch in word:
            if ch not in n.children:
                n.children[ch] = TrieNode()
            n = n.children[ch]
        n.is_word = True

    def search(self, word):
        n = self._walk(word)
        return n is not None and n.is_word

    def starts_with(self, prefix):
        return self._walk(prefix) is not None

    def _walk(self, s):
        n = self.root
        for ch in s:
            if ch not in n.children:
                return None
            n = n.children[ch]
        return n

    def autocomplete(self, prefix):
        """Gợi ý các từ bắt đầu bằng prefix."""
        n = self._walk(prefix)
        if n is None:
            return []
        out = []
        def dfs(node, path):
            if node.is_word:
                out.append(prefix + path)
            for ch, child in sorted(node.children.items()):
                dfs(child, path + ch)
        dfs(n, "")
        return sorted(out)


tu_dien = ["cat", "car", "card", "care", "dog", "do", "done"]
tr = Trie()
for w in tu_dien:
    tr.insert(w)

print("Từ điển:", tu_dien)
print("search('car')      :", tr.search("car"))
print("search('ca')       :", tr.search("ca"))
print("starts_with('ca')  :", tr.starts_with("ca"))
print("autocomplete('car'):", tr.autocomplete("car"))
print("autocomplete('do') :", tr.autocomplete("do"))
````

````python
# ---------- Ứng dụng 2: Heap (hàng đợi ưu tiên) ----------
class MinHeap:
    """Heap nhị phân tối thiểu, cài đặt thủ công (không dùng heapq)."""

    def __init__(self):
        self.a = []

    def push(self, x):
        self.a.append(x)
        self._up(len(self.a) - 1)

    def pop(self):
        if not self.a:
            raise IndexError("heap rỗng")
        top = self.a[0]
        last = self.a.pop()
        if self.a:
            self.a[0] = last
            self._down(0)
        return top

    def _up(self, i):
        while i > 0:
            p = (i - 1) // 2
            if self.a[p] <= self.a[i]:
                break
            self.a[p], self.a[i] = self.a[i], self.a[p]
            i = p

    def _down(self, i):
        n = len(self.a)
        while True:
            l, r, m = 2 * i + 1, 2 * i + 2, i
            if l < n and self.a[l] < self.a[m]: m = l
            if r < n and self.a[r] < self.a[m]: m = r
            if m == i:
                break
            self.a[i], self.a[m] = self.a[m], self.a[i]
            i = m


import heapq, random

# heap sort bằng heap tự viết
def heap_sort_mang(arr):
    h = MinHeap()
    for x in arr:
        h.push(x)
    return [h.pop() for _ in range(len(arr))]


test = [7, 3, 9, 1, 5, 8, 2]
print("Mảng      :", test)
print("Heap sort :", heap_sort_mang(test))
print("sorted()  :", sorted(test))
````

````python
# ---------- Ứng dụng 3: Cây biểu thức ----------
class ExprNode:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def eval_expr(node):
    """Tính giá trị cây biểu thức."""
    if node.left is None and node.right is None:
        return float(node.val)
    a, b = eval_expr(node.left), eval_expr(node.right)
    if node.val == '+': return a + b
    if node.val == '-': return a - b
    if node.val == '*': return a * b
    if node.val == '/': return a / b
    raise ValueError(node.val)


# Biểu thức (3 + 4) * 5  ->  cây biểu thức
expr = ExprNode('*', ExprNode('+', ExprNode('3'), ExprNode('4')), ExprNode('5'))

def in_order(node):
    """Duyệt giữa có ngoặc -> in lại biểu thức infix."""
    if node.left is None and node.right is None:
        return node.val
    return "(" + in_order(node.left) + node.val + in_order(node.right) + ")"

print("InOrder (infix):", in_order(expr))
print("Giá trị        :", eval_expr(expr), " (=(3+4)*5 = 35)")
````

````python
# ---- KIỂM THỬ ----
# Trie
assert tr.search("car") and tr.search("card") and not tr.search("ca")
assert tr.starts_with("ca") and not tr.starts_with("xy")
assert tr.autocomplete("car") == ["car", "card", "care"]
assert tr.autocomplete("do") == ["do", "dog", "done"]
print("Test Trie (search / starts_with / autocomplete): PASS")

# Heap: đối chiếu với heapq của Python
random.seed(42)
ok = True
for _ in range(200):
    arr = [random.randint(-50, 50) for _ in range(random.randint(0, 30))]
    if heap_sort_mang(arr) != sorted(arr):
        ok = False
        break
print("Test heap sort == sorted() với 200 mảng ngẫu nhiên:", "PASS" if ok else "FAIL")

# Heap: thứ tự pop luôn không giảm
h = MinHeap()
for x in [5, 1, 9, 3, 7, 1, 0]:
    h.push(x)
popped = [h.pop() for _ in range(7)]
assert popped == sorted([5, 1, 9, 3, 7, 1, 0]), popped
print("Test MinHeap pop ra dãy không giảm: PASS")

# Cây biểu thức
assert eval_expr(expr) == 35
assert in_order(expr) == "((3+4)*5)"
print("Test cây biểu thức (3+4)*5 = 35: PASS")

# Đối chiếu heap tự viết với heapq chuẩn (cùng dãy push/pop)
random.seed(7)
arr = [random.randint(-100, 100) for _ in range(50)]
h1 = MinHeap(); [h1.push(x) for x in arr]
h2 = []; [heapq.heappush(h2, x) for x in arr]
assert [h1.pop() for _ in range(50)] == [heapq.heappop(h2) for _ in range(50)]
print("Test MinHeap tự viết == heapq chuẩn: PASS")
````

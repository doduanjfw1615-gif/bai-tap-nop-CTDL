## Bài tập 1 (tr. 27, "Bạn tự viết") — Cấp số cộng

Tìm phần tử thứ $n$ của cấp số cộng có số hạng đầu $a$, công sai $r$:

$$a_1 = a,\qquad a_n = a_{n-1} + r$$

**Hệ thức truy hồi:** $\text{CSC}(n) = \begin{cases} a & n = 1\\ r + \text{CSC}(n-1) & n > 1\end{cases}$

```python
def tinh_cap_so_cong(n, a, r):
    """Số hạng thứ n của cấp số cộng (đệ quy tuyến tính)."""
    if n == 1:                 # trường hợp suy biến (neo)
        return a
    return r + tinh_cap_so_cong(n - 1, a, r)


def tinh_cap_so_cong_vong(n, a, r):
    """Cùng bài toán, viết bằng vòng lặp (khử đệ quy)."""
    kq = a
    for _ in range(n - 1):
        kq += r
    return kq


# --- Minh họa theo đúng ví dụ trong slide: a = 3, r = 4 ---
a, r = 3, 4
print("a = 3, r = 4:")
for n in range(1, 6):
    print(f"  a{n} = {tinh_cap_so_cong(n, a, r)}")
```

```python
# ---- KIỂM THỬ ----
# 1) Kiểm tra theo công thức đóng của cấp số cộng: a_n = a + (n-1)*r
ok = all(tinh_cap_so_cong(n, a, r) == a + (n - 1) * r for n in range(1, 200))
print("Test công thức đóng a + (n-1)r với n=1..199:", "PASS" if ok else "FAIL")

# 2) Đối chiếu bản đệ quy với bản vòng lặp
ok2 = all(tinh_cap_so_cong(n, 7, -3) == tinh_cap_so_cong_vong(n, 7, -3) for n in range(1, 100))
print("Test đệ quy == vòng lặp (a=7, r=-3):", "PASS" if ok2 else "FAIL")

# 3) Giá trị cụ thể đã biết trong slide
assert tinh_cap_so_cong(1, 3, 4) == 3
assert tinh_cap_so_cong(5, 3, 4) == 19
print("Test giá trị slide (a1=3, a5=19): PASS")
```

## Bài tập 2 (tr. 29, "Bạn tự viết") — Biểu diễn nhị phân

Xuất biểu diễn nhị phân của số nguyên dương $n$. Ví dụ $13 \to 1101$.

**Giải thuật:** nếu $n \neq 0$ thì in biểu diễn nhị phân của $\lfloor n/2 \rfloor$ rồi in $n \bmod 2$.

Thứ tự **đệ quy trước, in sau** chính là mấu chốt: phép gọi đệ quy chạy trước nên các bit được in từ bit cao (trái) đến bit thấp (phải) — đúng chiều đọc.

```python
def bieu_dien_nhi_phan(n):
    """In biểu diễn nhị phân của n (đệ quy tuyến tính)."""
    if n != 0:
        bieu_dien_nhi_phan(n // 2)   # đệ quy TRƯỚC
        print(n % 2, end='')         # in bit SAU -> bit cao in trước


def np_string(n):
    """Trả về chuỗi nhị phân (tiện cho việc kiểm thử).

    Mấu chốt: trường hợp cơ sở trả về chuỗi RỖNG (không phải '0'),
    nếu không sẽ sinh thêm một bit 0 ở đầu (13 -> '01101').
    """
    if n == 0:
        return ""
    return np_string(n // 2) + str(n % 2)


print("13  ->", end=' '); bieu_dien_nhi_phan(13); print()
print("10  ->", end=' '); bieu_dien_nhi_phan(10); print()
print("255 ->", end=' '); bieu_dien_nhi_phan(255); print()
```

```python
# ---- KIỂM THỬ ----
# 1) Đối chiếu với hàm nhị phân có sẵn của Python
ok = all(np_string(n) == bin(n)[2:] for n in range(1, 5000))
print("Test np_string(n) == bin(n)[2:] với n=1..4999:", "PASS" if ok else "FAIL")

# 2) Ví dụ trong slide
assert np_string(13) == "1101"
print("Test 13 -> 1101: PASS")

# 3) Số 0 (trường hợp biên)
print("np_string(0) =", np_string(0), "(bin(0)[2:] =", bin(0)[2:], ")")
```

## Bài tập 3 (tr. 52–53) — Ví dụ 9: Tính giá trị đa thức

Dãy $\{X_n\}$ cho bởi:

$$X_0 = 1,\qquad X_n = n^2 X_0 + (n-1)^2 X_1 + \dots + 2^2 X_{n-2} + 1^2 X_{n-1} = \sum_{i=0}^{n-1} (n-i)^2 X_i$$

Đây là **đệ quy phi tuyến**: lời gọi đệ quy nằm trong vòng lặp, mỗi bài toán kích thước $n$ gọi $n$ bài toán con nhỏ hơn.

**Phân tích độ phức tạp:** $T(n) = T(n-1) + T(n-2) + \dots + T(0) + O(n)$. Với $T(0)=1$ ta có $T(n) = 2T(n-1) + O(1)$, suy ra $T(n) = O(2^n)$ — bùng nổ theo hàm mũ.

```python
def tinh_X(n):
    """X_n = sum_{i=0}^{n-1} (n-i)^2 * X_i  (đệ quy phi tuyến, đúng theo slide)."""
    if n == 0:
        return 1
    tong = 0
    for i in range(n):
        tong += (n - i) * (n - i) * tinh_X(i)
    return tong


def tinh_X_dp(n):
    """Bản quy hoạch động: tính từ dưới lên, mỗi X_i chỉ tính 1 lần -> O(n^2)."""
    X = [1] * (n + 1)
    for k in range(1, n + 1):
        X[k] = sum((k - i) * (k - i) * X[i] for i in range(k))
    return X[n]


print("Đệ quy phi tuyến (bản slide):")
for n in range(0, 8):
    print(f"  X{n} = {tinh_X(n)}")
print("\nQuy hoạch động (bản tối ưu):")
for n in range(0, 8):
    print(f"  X{n} = {tinh_X_dp(n)}")
```

```python
# ---- KIỂM THỬ ----
# 1) Hai bản cài đặt phải cho cùng kết quả
ok = all(tinh_X(n) == tinh_X_dp(n) for n in range(0, 15))
print("Test đệ quy == quy hoạch động với n=0..14:", "PASS" if ok else "FAIL")

# 2) Kiểm tra trực tiếp định nghĩa (tính tổng bằng tay, độc lập)
X = [1]
for n in range(1, 15):
    X.append(sum((n - i) ** 2 * X[i] for i in range(n)))
ok2 = all(tinh_X_dp(n) == X[n] for n in range(0, 15))
print("Test theo định nghĩa (tính tổng tường minh):", "PASS" if ok2 else "FAIL")

# 3) Trường hợp cơ sở
assert tinh_X(0) == 1 and tinh_X_dp(0) == 1
print("Test X0 = 1: PASS")
```

## Bài tập 4 (tr. 58) — Bài toán Chuyển đĩa (Tháp Hà Nội)

Chuyển $N$ đĩa từ cột 1 sang cột 2, dùng cột 3 làm trung gian. Mỗi lần chuyển 1 đĩa, đĩa nhỏ luôn nằm trên đĩa lớn.

**Giải thuật đệ quy:**
1. Chuyển $N-1$ đĩa từ cột nguồn sang cột trung gian.
2. Chuyển đĩa lớn nhất từ cột nguồn sang cột đích.
3. Chuyển $N-1$ đĩa từ cột trung gian sang cột đích.

**Độ phức tạp:** $T(n) = 2T(n-1) + 1$, $T(1)=1 \Rightarrow T(n) = 2^n - 1$ bước.

```python
def thap_ha_noi(n, nguon, dich, trung_gian, buoc=None):
    """Chuyển n đĩa từ cột `nguon` sang cột `dich`, in từng bước."""
    if buoc is None:
        buoc = []
    if n == 1:
        buoc.append((nguon, dich))
        return buoc
    thap_ha_noi(n - 1, nguon, trung_gian, dich, buoc)   # (1) n-1 đĩa sang trung gian
    buoc.append((nguon, dich))                          # (2) đĩa lớn nhất sang đích
    thap_ha_noi(n - 1, trung_gian, dich, nguon, buoc)   # (3) n-1 đĩa về đích
    return buoc


def kiem_tra_loi_giai(buoc, n):
    """Mô phỏng lại các bước chuyển đĩa, kiểm tra luôn hợp lệ (đĩa nhỏ trên đĩa lớn)."""
    cot = {1: list(range(n, 0, -1)), 2: [], 3: []}   # mỗi cột: list, phần tử cuối là đĩa trên cùng
    for (tu, den) in buoc:
        if not cot[tu]:
            return False, f"Chuyển từ cột {tu} nhưng cột rỗng"
        d = cot[tu].pop()
        if cot[den] and cot[den][-1] < d:
            return False, f"Đặt đĩa {d} lên đĩa nhỏ hơn {cot[den][-1]}"
        cot[den].append(d)
    if cot[2] != list(range(n, 0, -1)):
        return False, "Kết thúc: cột 2 không chứa đủ đĩa theo đúng thứ tự"
    return True, "OK"


for n in (1, 2, 3):
    buoc = thap_ha_noi(n, 1, 2, 3)
    print(f"N = {n}: {len(buoc)} bước")
    for i, (a_, b_) in enumerate(buoc, 1):
        print(f"   bước {i}: cột {a_} -> cột {b_}")
    print()
```

```python
# ---- KIỂM THỬ ----
# 1) Số bước phải đúng bằng 2^n - 1
ok = all(len(thap_ha_noi(n, 1, 2, 3)) == 2 ** n - 1 for n in range(1, 13))
print("Test số bước == 2^n - 1 (n=1..12):", "PASS" if ok else "FAIL")

# 2) Mô phỏng lại từng bước: kiểm tra quy tắc "đĩa nhỏ trên đĩa lớn"
allok = True
for n in range(1, 11):
    hop_le, msg = kiem_tra_loi_giai(thap_ha_noi(n, 1, 2, 3), n)
    if not hop_le:
        allok = False
        print(f"  N={n} LỖI: {msg}")
print("Test mô phỏng hợp lệ cho n=1..10:", "PASS" if allok else "FAIL")

# 3) Trường hợp cụ thể N=3 (7 bước)
assert len(thap_ha_noi(3, 1, 2, 3)) == 7
print("Test N=3 có 7 bước: PASS")
```

## Bài tập 5 (tr. 64) — Phương pháp Thay thế

**Đề:** Cho hệ thức truy hồi $T(n) = 2T(n/2) + n$ với $T(1) = 1$. Dùng **phương pháp thay thế**, chứng minh $T(n) = O(n \log n)$.

### Lời giải

**Bước 1 — Đoán nghiệm.** Đoán $T(n) \le c\,n\log n$ với hằng số $c$ đủ lớn (dùng $\log_2$).

**Bước 2 — Chứng minh bằng quy nạp.** Giả sử $T(k) \le c\,k\log k$ đúng với mọi $k < n$. Khi đó:

$$\begin{aligned}
T(n) &= 2T(n/2) + n \\
     &\le 2\Big(c\,\tfrac{n}{2}\log\tfrac{n}{2}\Big) + n \\
     &= c\,n(\log n - 1) + n \\
     &= c\,n\log n - (c-1)n \\
     &\le c\,n\log n \quad \text{khi } c \ge 1
\end{aligned}$$

**Bước 3 — Trường hợp cơ sở.** Với $n=1$: $c\cdot 1\cdot\log 1 = 0 < T(1) = 1$ — **không đúng**. Xét $n = 2$: $T(2) = 2T(1) + 2 = 4$, cần $4 \le c\cdot 2\log 2 = 2c \Rightarrow c \ge 2$.

**Kết luận:** chọn $c = 2$ (và điều kiện $c \ge 1$ đã thoả), ta được $T(n) \le 2n\log n$, tức

$$\boxed{T(n) = O(n\log n)}$$

Phần code dưới đây kiểm chứng lại chặn $T(n) \le 2n\log_2 n$ bằng số liệu thực nghiệm.

```python
import math

def T(n):
    """T(n) = 2T(n/2) + n, T(1) = 1 (n là luỹ thừa của 2)."""
    if n == 1:
        return 1
    return 2 * T(n // 2) + n


print(f"{'n':>8} {'T(n)':>12} {'2*n*log2(n)':>14} {'T <= 2nlog2n?':>14}")
ok = True
for k in range(0, 16):
    n = 2 ** k
    lhs, rhs = T(n), 2 * n * math.log2(n) if n > 1 else 2 * n * 0
    if n > 1 and lhs > rhs:
        ok = False
    print(f"{n:>8} {lhs:>12} {rhs:>14.1f} {str(lhs <= rhs):>14}")
print("\nChặn T(n) <= 2n·log2(n) đúng với mọi n đã xét:", "PASS" if ok else "FAIL")
```

```python
# ---- KIỂM THỬ ----
# 1) Kiểm tra nghiệm đóng của hệ thức: T(n) = n*log2(n) + n  (với n = 2^k)
def T_dong(n):
    return n * math.log2(n) + n if n > 1 else 1

ok = all(abs(T(2 ** k) - T_dong(2 ** k)) < 1e-9 for k in range(0, 20))
print("Test T(n) == n·log2(n) + n cho n = 2^k (k=0..19):", "PASS" if ok else "FAIL")

# 2) Kiểm tra T(n) = Theta(n log n): tỉ số T(n)/(n log n) bị chặn hai đầu
ratio = [T(2 ** k) / (2 ** k * math.log2(2 ** k)) for k in range(2, 20)]
print(f"Tỉ số T(n)/(n·log2 n) min={min(ratio):.4f} max={max(ratio):.4f} -> nằm trong [1, 2]")
print("Test Theta(n log n):", "PASS" if 1 <= min(ratio) and max(ratio) <= 2 else "FAIL")

# 3) Hằng số c=2 trong chứng minh: T(2) = 4 <= 2*2*log2(2) = 4
assert T(2) == 4 and 2 * 2 * math.log2(2) == 4
print("Test trường hợp cơ sở n=2 với c=2 (T(2)=4 <= 4): PASS")
```

## Bài tập 6 (tr. 68) — Phương pháp Cây Đệ quy

**Đề:** Dùng phương pháp **cây đệ quy** tìm chặn tiệm cận $\Theta$ cho $T(n) = 2T(n/4) + \sqrt{n}$, với $T(1) = 1$.

### Lời giải

**Mức 0 (gốc):** chi phí $\sqrt{n}$, 1 nút.

**Mức 1:** 2 nút, mỗi nút kích thước $n/4$, chi phí mỗi nút $\sqrt{n/4} = \tfrac{1}{2}\sqrt{n}$.
Tổng mức 1: $2 \cdot \tfrac{1}{2}\sqrt{n} = \sqrt{n}$.

**Mức 2:** $2^2$ nút, kích thước $n/4^2$, chi phí mỗi nút $\sqrt{n/4^2} = \tfrac{1}{4}\sqrt{n}$.
Tổng mức 2: $4 \cdot \tfrac{1}{4}\sqrt{n} = \sqrt{n}$.

**Mức $k$ (tổng quát):** $2^k$ nút, kích thước $n/4^k$, mỗi nút $\sqrt{n}/2^k$.
Tổng mức $k$: $2^k \cdot \dfrac{\sqrt{n}}{2^k} = \sqrt{n}$ — **mỗi mức đều tốn $\sqrt{n}$**.

**Chiều cao cây:** dừng khi $n/4^h = 1 \Rightarrow h = \log_4 n = \tfrac{1}{2}\log_2 n$. Số mức là $h + 1$.

**Tổng chi phí các mức trong:**
$$\sum_{k=0}^{h} \sqrt{n} = \sqrt{n}\,(\log_4 n + 1) = \Theta(\sqrt{n}\log n)$$

**Chi phí ở lá:** số lá $2^h = 2^{\log_4 n} = n^{1/2} = \sqrt{n}$, mỗi lá $T(1)=1$
$\Rightarrow \Theta(\sqrt{n})$.

**Kết luận:**
$$\boxed{T(n) = \Theta(\sqrt{n}\log n) + \Theta(\sqrt{n}) = \Theta(\sqrt{n}\log n)}$$

```python
def T2(n):
    """T(n) = 2T(n/4) + sqrt(n), T(1) = 1 (n là luỹ thừa của 4)."""
    if n == 1:
        return 1
    return 2 * T2(n // 4) + n ** 0.5


def T2_tong_muc(n):
    """Công thức cây đệ quy: h mức trong, mỗi mức sqrt(n), cộng sqrt(n) ở lá
       => sqrt(n)*(h+1) với h = log4(n)."""
    h = math.log(n, 4)
    return n ** 0.5 * (h + 1)


print(f"{'n':>10} {'T(n)':>14} {'sqrt(n)*(log4 n+1)+sqrt(n)':>30}")
for k in range(0, 9):
    n = 4 ** k
    print(f"{n:>10} {T2(n):>14.4f} {T2_tong_muc(n):>30.4f}")
```

```python
# ---- KIỂM THỬ ----
# 1) Công thức cây đệ quy khớp với giá trị truy hồi thực tế
ok = all(abs(T2(4 ** k) - T2_tong_muc(4 ** k)) < 1e-6 for k in range(0, 12))
print("Test T(n) == sqrt(n)(log4 n + 1) + sqrt(n) với n = 4^k (k=0..11):", "PASS" if ok else "FAIL")

# 2) Kiểm tra Theta(sqrt(n) log n): tỉ số T(n)/(sqrt(n)*log2 n) bị chặn 2 đầu
ratio = [T2(4 ** k) / ((4 ** k) ** 0.5 * math.log2(4 ** k)) for k in range(2, 12)]
print(f"Tỉ số T(n)/(sqrt(n)·log2 n): min={min(ratio):.4f} max={max(ratio):.4f}")
print("Test Theta(sqrt(n) log n):", "PASS" if max(ratio) / min(ratio) < 2 else "FAIL")

# 3) Số lá của cây đúng bằng sqrt(n)
ok3 = all(2 ** math.log(4 ** k, 4) == (4 ** k) ** 0.5 for k in range(0, 12))
print("Test số lá 2^h == sqrt(n):", "PASS" if ok3 else "FAIL")
```

## Luyện tập thêm (theo phần "Thuật toán quay lui", tr. 71–76)

Slide giới thiệu **backtracking** với ví dụ hoán vị và các ứng dụng (n quân hậu, Sudoku, tổ hợp, mê cung, Knapsack). Dưới đây cài đặt và **kiểm thử** hai bài kinh điển.

```python
def hoan_vi(nums):
    """Liệt kê tất cả hoán vị của nums bằng quay lui (đúng theo code trong slide)."""
    result = []
    used = [False] * len(nums)

    def backtrack(path):
        if len(path) == len(nums):
            result.append(path[:])
            return
        for i in range(len(nums)):
            if not used[i]:
                used[i] = True
                path.append(nums[i])
                backtrack(path)
                path.pop()          # quay lui
                used[i] = False

    backtrack([])
    return result


def n_quan_hau(n):
    """Đếm số cách đặt n quân hậu trên bàn n x n (quay lui)."""
    dem = 0
    cot = set(); cheo_xuoi = set(); cheo_nguoc = set()

    def dat(hang):
        nonlocal dem
        if hang == n:
            dem += 1
            return
        for c in range(n):
            if c in cot or (hang - c) in cheo_xuoi or (hang + c) in cheo_nguoc:
                continue
            cot.add(c); cheo_xuoi.add(hang - c); cheo_nguoc.add(hang + c)
            dat(hang + 1)
            cot.remove(c); cheo_xuoi.remove(hang - c); cheo_nguoc.remove(hang + c)

    dat(0)
    return dem


print("Hoán vị của [1,2,3]:")
for p in hoan_vi([1, 2, 3]):
    print("  ", p)
```

```python
# ---- KIỂM THỬ ----
import itertools

# 1) Hoán vị: số lượng = n!, và tập hợp phải trùng với itertools.permutations
for n in range(1, 7):
    arr = list(range(1, n + 1))
    got = sorted(hoan_vi(arr))
    exp = sorted(list(p) for p in itertools.permutations(arr))
    assert got == exp, f"hoán vị sai với n={n}"
print("Test hoán vị == itertools.permutations với n=1..6: PASS")

# 2) n quân hậu: đối chiếu với dãy số đã biết (OEIS A000170)
#    1, 0, 0, 2, 10, 4, 40, 92, 352
expect = {1: 1, 2: 0, 3: 0, 4: 2, 5: 10, 6: 4, 7: 40, 8: 92}
ok = True
for n, e in expect.items():
    got = n_quan_hau(n)
    if got != e:
        ok = False
        print(f"  n={n}: got {got}, expect {e}")
print("Test n quân hậu đối chiếu dãy chuẩn (n=1..8):", "PASS" if ok else "FAIL")

# 3) Kiểm tra nghiệm n=4 có đúng 2 cách (bài toán kinh điển)
assert n_quan_hau(4) == 2
print("Test 4 quân hậu = 2 nghiệm: PASS")
```

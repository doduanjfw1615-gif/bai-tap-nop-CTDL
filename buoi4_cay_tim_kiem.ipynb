## BT1 (tr. 18) — Vẽ BST và liệt kê 4 thứ tự duyệt

Dãy nhập từ trái sang phải: `15; 30; 46; 3; 21; 7; 9; 10; 29; 34; 13; 5; 37; 18`

```python
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


class BST:
    """Cây nhị phân tìm kiếm với đầy đủ thao tác."""

    def __init__(self):
        self.root = None

    # ---------------- Cơ bản ----------------
    def is_empty(self):
        return self.root is None

    def insert(self, key):
        if self.root is None:
            self.root = Node(key)
            return True
        return self._insert(self.root, key)

    def _insert(self, node, key):
        if key == node.key:
            return False
        if key < node.key:
            if node.left is None:
                node.left = Node(key)
                return True
            return self._insert(node.left, key)
        if node.right is None:
            node.right = Node(key)
            return True
        return self._insert(node.right, key)

    def build(self, keys):
        for k in keys:
            self.insert(k)
        return self

    def search(self, key):
        return self._search(self.root, key)

    def _search(self, node, key):
        if node is None or node.key == key:
            return node
        return self._search(node.left, key) if key < node.key else self._search(node.right, key)

    # ---------------- Duyệt ----------------
    def preorder(self):
        out = []; self._pre(self.root, out); return out

    def _pre(self, n, out):
        if n:
            out.append(n.key); self._pre(n.left, out); self._pre(n.right, out)

    def inorder(self):
        out = []; self._in(self.root, out); return out

    def _in(self, n, out):
        if n:
            self._in(n.left, out); out.append(n.key); self._in(n.right, out)

    def postorder(self):
        out = []; self._post(self.root, out); return out

    def _post(self, n, out):
        if n:
            self._post(n.left, out); self._post(n.right, out); out.append(n.key)

    def breadth_first(self):
        if self.root is None:
            return []
        out, q = [], [self.root]
        while q:
            n = q.pop(0)
            out.append(n.key)
            if n.left: q.append(n.left)
            if n.right: q.append(n.right)
        return out

    # ---------------- Vẽ ----------------
    def levels(self):
        if self.root is None:
            return []
        res, cur = [], [self.root]
        while any(cur):
            res.append([n.key if n else None for n in cur])
            nxt = []
            for n in cur:
                nxt.append(n.left if n else None)
                nxt.append(n.right if n else None)
            cur = nxt
        return res

    def draw(self):
        lv = self.levels()
        if not lv:
            print("(cây rỗng)"); return
        for i, row in enumerate(lv):
            cell = 2 ** (len(lv) - i)
            print("".join(str(v).center(cell) if v is not None else " " * cell for v in row).rstrip())
        print()


day1 = [15, 30, 46, 3, 21, 7, 9, 10, 29, 34, 13, 5, 37, 18]
t1 = BST().build(day1)
print("Dãy nhập:", day1, "\n")
print("Cây BST (BT1):")
t1.draw()
print("PreOrder  (NLR):", t1.preorder())
print("InOrder   (LNR):", t1.inorder())
print("PostOrder (LRN):", t1.postorder())
print("Breadth-first  :", t1.breadth_first())
```

```python
# ---- KIỂM THỬ BT1 ----
assert t1.inorder() == sorted(day1), "InOrder phải là dãy tăng"
assert t1.breadth_first() == [v for r in t1.levels() for v in r if v is not None]
assert sorted(t1.preorder()) == sorted(day1)
assert sorted(t1.postorder()) == sorted(day1)

# kiểm tra tính chất BST
def ktra(node, lo=float('-inf'), hi=float('inf')):
    if node is None: return True
    return lo < node.key < hi and ktra(node.left, lo, node.key) and ktra(node.right, node.key, hi)
assert ktra(t1.root)
print("Test BT1 (InOrder tăng, BFS theo mức, tính chất BST): PASS")
```

## BT2 (tr. 26) & BT3 (tr. 27) — Tạo cây, duyệt, tìm kiếm, xóa nút

**BT2:** Tạo cây từ bàn phím (dừng khi gặp giá trị trùng) · duyệt LNR/LRN/NLR · tìm kiếm · xóa nút bằng **Copying** và **Merging**.

**BT3:** Tạo cây từ bàn phím (dừng khi trùng *hoặc hết bộ nhớ*) · tạo cây từ dãy cho trước.

### Hai phương pháp xóa nút có 2 cây con

| Phương pháp | Cách làm |
|---|---|
| **Copying** | Tìm nút thay thế = **trái nhất của cây con phải** (hoặc phải nhất của cây con trái), **chép giá trị** vào nút cần xóa, rồi xóa nút thay thế |
| **Merging** | **Nối (merge)** cây con trái và cây con phải của nút cần xóa thành một cây, rồi gắn cây đó vào vị trí nút cần xóa (không chép giá trị) |

```python
class BSTDelete(BST):
    """Bổ sung 2 phương pháp xóa nút: Copying và Merging."""

    # ---------- XÓA theo phương pháp COPYING ----------
    def delete_copying(self, key):
        """Xóa nút chứa key bằng cách chép giá trị nút thay thế."""
        self.root = self._del_copy(self.root, key)
        return True

    def _del_copy(self, node, key):
        if node is None:
            return None
        if key < node.key:
            node.left = self._del_copy(node.left, key)
        elif key > node.key:
            node.right = self._del_copy(node.right, key)
        else:
            # nút lá hoặc nút có 1 cây con
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left
            # nút có 2 cây con: tìm trái nhất của cây con phải
            thay_the = node.right
            while thay_the.left is not None:
                thay_the = thay_the.left
            node.key = thay_the.key                       # CHÉP giá trị
            node.right = self._del_copy(node.right, thay_the.key)
        return node

    # ---------- XÓA theo phương pháp MERGING ----------
    def delete_merging(self, key):
        """Xóa nút chứa key bằng cách nối (merge) 2 cây con lại."""
        self.root = self._del_merge(self.root, key)
        return True

    def _del_merge(self, node, key):
        if node is None:
            return None
        if key < node.key:
            node.left = self._del_merge(node.left, key)
            return node
        if key > node.key:
            node.right = self._del_merge(node.right, key)
            return node
        # tìm thấy nút cần xóa
        if node.left is None:
            return node.right
        if node.right is None:
            return node.left
        # 2 cây con: nối cây con phải vào chỗ trống phải nhất của cây con trái
        trai, phai = node.left, node.right
        cur = trai
        while cur.right is not None:
            cur = cur.right
        cur.right = phai                                  # MERGE
        return trai

    # ---------- Tìm min / max ----------
    def find_min(self, node=None, first=True):
        if first:
            node = self.root
        if node is None:
            return None
        while node.left is not None:
            node = node.left
        return node.key

    def find_max(self, node=None, first=True):
        if first:
            node = self.root
        if node is None:
            return None
        while node.right is not None:
            node = node.right
        return node.key


# BT3: tạo cây từ dãy cho trước
day_bt3 = [50, 30, 70, 20, 40, 60, 80, 35, 45]
t3 = BSTDelete().build(day_bt3)
print("BT3 - Cây từ dãy", day_bt3, ":")
t3.draw()
print("InOrder:", t3.inorder())
```

```python
# Minh họa xóa 2 phương pháp trên cùng một cây
import copy

goc = BSTDelete().build([50, 30, 70, 20, 40, 60, 80, 35, 45])

for pp in ("copying", "merging"):
    t = BSTDelete()
    t.root = copy.deepcopy(goc.root)
    print(f"=== Xóa 30 bằng phương pháp {pp.upper()} ===")
    print("Trước:", t.inorder())
    if pp == "copying":
        t.delete_copying(30)
    else:
        t.delete_merging(30)
    print("Sau  :", t.inorder())
    t.draw()
```

```python
# ---- KIỂM THỬ BT2/BT3 ----
# 1) Xóa bằng cả 2 phương pháp đều giữ đúng tập giá trị còn lại và tính chất BST
goc_keys = [50, 30, 70, 20, 40, 60, 80, 35, 45]
for pp in ("copying", "merging"):
    for x in goc_keys:
        t = BSTDelete().build(goc_keys)
        (t.delete_copying if pp == "copying" else t.delete_merging)(x)
        con_lai = sorted(k for k in goc_keys if k != x)
        assert t.inorder() == con_lai, f"{pp}: xóa {x} sai -> {t.inorder()}"
        assert ktra(t.root), f"{pp}: xóa {x} làm hỏng tính chất BST"
print("Test xóa mọi nút bằng Copying và Merging (9 nút x 2 pp): PASS")

# 2) Xóa nút không tồn tại -> cây không đổi
t = BSTDelete().build(goc_keys)
t.delete_copying(999); t.delete_merging(999)
assert t.inorder() == sorted(goc_keys)
print("Test xóa giá trị không tồn tại: PASS")

# 3) Xóa hết cây dần dần -> rỗng
t = BSTDelete().build(goc_keys)
for x in goc_keys:
    t.delete_copying(x)
assert t.is_empty()
print("Test xóa lần lượt đến khi cây rỗng: PASS")

# 4) min/max
t = BSTDelete().build(goc_keys)
assert t.find_min() == min(goc_keys) and t.find_max() == max(goc_keys)
print("Test find_min/find_max: PASS")
```

## BT4, BT5, BT6 (tr. 32, 35, 38) — Duyệt cây BST theo thứ tự trước / giữa / sau

Ba bài tập này yêu cầu viết hàm duyệt **NLR**, **LNR**, **LRN** và minh họa từng bước. Dưới đây cài đặt cả bản **đệ quy** (đúng theo slide) và bản **không đệ quy dùng stack** (khử đệ quy), rồi đối chiếu kết quả.

Slide tr. 40 còn nêu cách **kiểm tra kết quả duyệt bằng cách dựng lại cây**: từ kết quả NLR chọn giá trị đầu làm gốc; từ kết quả LRN chọn giá trị cuối làm gốc — notebook thực hiện luôn phép kiểm tra này.

```python
def preorder_stack(root):
    """Duyệt NLR không đệ quy (dùng stack)."""
    if root is None:
        return []
    out, st = [], [root]
    while st:
        n = st.pop()
        out.append(n.key)
        if n.right: st.append(n.right)
        if n.left:  st.append(n.left)
    return out


def inorder_stack(root):
    """Duyệt LNR không đệ quy (dùng stack)."""
    out, st, cur = [], [], root
    while st or cur:
        while cur:
            st.append(cur)
            cur = cur.left
        cur = st.pop()
        out.append(cur.key)
        cur = cur.right
    return out


def postorder_stack(root):
    """Duyệt LRN không đệ quy (dùng 2 stack)."""
    if root is None:
        return []
    out, st1, st2 = [], [root], []
    while st1:
        n = st1.pop()
        st2.append(n)
        if n.left:  st1.append(n.left)
        if n.right: st1.append(n.right)
    while st2:
        out.append(st2.pop().key)
    return out


print("BT4 - PreOrder (NLR), đệ quy  :", t1.preorder())
print("BT4 - PreOrder (NLR), dùng stack:", preorder_stack(t1.root))
print()
print("BT5 - InOrder (LNR), đệ quy   :", t1.inorder())
print("BT5 - InOrder (LNR), dùng stack:", inorder_stack(t1.root))
print()
print("BT6 - PostOrder (LRN), đệ quy :", t1.postorder())
print("BT6 - PostOrder (LRN), dùng stack:", postorder_stack(t1.root))
```

```python
# ---- KIỂM THỬ BT4/BT5/BT6 ----
assert preorder_stack(t1.root) == t1.preorder()
assert inorder_stack(t1.root) == t1.inorder()
assert postorder_stack(t1.root) == t1.postorder()
print("Test 3 cách duyệt: bản đệ quy == bản dùng stack: PASS")

# Dựng lại cây từ kết quả duyệt (cách kiểm tra ở slide tr.40):
#   NLR -> phần tử đầu là gốc; LRN -> phần tử cuối là gốc
def rebuild_from_preorder(keys):
    t = BST()
    for k in keys:
        t.insert(k)
    return t

def rebuild_from_postorder(keys):
    t = BST()
    for k in reversed(keys):
        t.insert(k)
    return t

assert rebuild_from_preorder(t1.preorder()).inorder() == t1.inorder()
assert rebuild_from_postorder(t1.postorder()).inorder() == t1.inorder()
print("Test dựng lại cây từ kết quả NLR và LRN (slide tr.40): PASS")

# Duyệt NLR của cây dựng lại trùng với kết quả gốc
assert rebuild_from_preorder(t1.preorder()).preorder() == t1.preorder()
assert rebuild_from_postorder(t1.postorder()).postorder() == t1.postorder()
print("Test cây dựng lại cho đúng kết quả duyệt gốc: PASS")
```

## BT7 (tr. 43) — Tìm kiếm trên BST

Cho BST mỗi node có giá trị nguyên, định nghĩa các hàm:
1. Tìm node có giá trị `x`
2. Tìm node có giá trị **lớn nhất**
3. Tìm node có giá trị **nhỏ nhất của cây con phải**

```python
class BSTSearch(BST):
    """Các hàm tìm kiếm của BT7."""

    def find_node(self, x):
        """1. Trả về node có giá trị x (None nếu không có)."""
        return self.search(x)

    def find_max_node(self):
        """2. Node lớn nhất: đi phải mãi (slide: 'max nằm ở nút phải nhất')."""
        if self.root is None:
            return None
        n = self.root
        while n.right is not None:
            n = n.right
        return n

    def find_min_node(self):
        """Node nhỏ nhất: đi trái mãi ('min nằm ở nút trái nhất')."""
        if self.root is None:
            return None
        n = self.root
        while n.left is not None:
            n = n.left
        return n

    def find_min_right_subtree(self, x):
        """3. Node nhỏ nhất của CÂY CON PHẢI của node giá trị x."""
        n = self.search(x)
        if n is None or n.right is None:
            return None
        m = n.right
        while m.left is not None:
            m = m.left
        return m

    def find_max_left_subtree(self, x):
        """(bổ sung) Node lớn nhất của CÂY CON TRÁI của node giá trị x."""
        n = self.search(x)
        if n is None or n.left is None:
            return None
        m = n.left
        while m.right is not None:
            m = m.right
        return m


ts = BSTSearch().build([7, 3, 36, 1, 6, 15, 40, 4, 23])   # đúng cây minh họa trong slide
print("Cây minh họa của slide:")
ts.draw()
print("Tìm x = 23          :", (ts.find_node(23) or type('x',(),{'key':None})).key)
print("Node lớn nhất       :", ts.find_max_node().key)
print("Node nhỏ nhất       :", ts.find_min_node().key)
print("Min cây con phải của 7  :", ts.find_min_right_subtree(7).key)
print("Max cây con trái của 7  :", ts.find_max_left_subtree(7).key)
print("Min cây con phải của 36 :", ts.find_min_right_subtree(36).key)
```

```python
# ---- KIỂM THỬ BT7 ----
day_ts = [7, 3, 36, 1, 6, 15, 40, 4, 23]

# 1) find_node
assert all(ts.find_node(x) is not None and ts.find_node(x).key == x for x in day_ts)
assert ts.find_node(999) is None
print("Test find_node (có / không có): PASS")

# 2) max = phần tử lớn nhất, min = phần tử nhỏ nhất
assert ts.find_max_node().key == max(day_ts) == 40
assert ts.find_min_node().key == min(day_ts) == 1
print("Test max = 40, min = 1: PASS")

# 3) min của cây con phải = min của các giá trị > x trong cây
for x in day_ts:
    n = ts.find_node(x)
    if n.right is not None:
        gia_tri_phai = [k for k in day_ts if k > x]
        assert ts.find_min_right_subtree(x).key == min(gia_tri_phai), f"sai tại x={x}"
print("Test min cây con phải == min các giá trị lớn hơn x (mọi x): PASS")

# 4) max của cây con trái = max các giá trị nhỏ hơn x
for x in day_ts:
    n = ts.find_node(x)
    if n.left is not None:
        gia_tri_trai = [k for k in day_ts if k < x]
        assert ts.find_max_left_subtree(x).key == max(gia_tri_trai), f"sai tại x={x}"
print("Test max cây con trái == max các giá trị nhỏ hơn x (mọi x): PASS")
```

## BT8 (tr. 44) — 10 hàm thông tin của cây

Cho BST mỗi node có giá trị nguyên, định nghĩa các hàm:
1. In các node có giá trị **chẵn**
2. Đếm số node của cây
3. Tính **độ cao** của cây
4. Đếm số node **lá** (node bậc 0)
5. Đếm số node có **1 cây con** (node bậc 1)
6. Đếm số node chỉ có **1 cây con phải**
7. Đếm số node có **1 cây con trái**
8. Đếm số node **2 cây con** (node bậc 2)
9. In các node **trên từng mức** của cây
10. Cho biết **độ dài đường đi từ gốc đến node x**

```python
class BSTInfo(BST):
    """10 hàm thông tin của cây (BT8)."""

    def _walk(self, node, fn):
        if node is None:
            return
        fn(node)
        self._walk(node.left, fn)
        self._walk(node.right, fn)

    # 1
    def print_even(self):
        out = []
        self._walk(self.root, lambda n: out.append(n.key) if n.key % 2 == 0 else None)
        print("Các node chẵn:", out)
        return out

    # 2
    def count_nodes(self):
        def dem(n):
            return 0 if n is None else 1 + dem(n.left) + dem(n.right)
        return dem(self.root)

    # 3
    def height(self, node=None, first=True):
        """Số mức của cây (gốc = mức 1). Cây rỗng -> 0."""
        if first:
            node = self.root
        if node is None:
            return 0
        return 1 + max(self.height(node.left, False), self.height(node.right, False))

    # 4
    def count_leaves(self, node=None, first=True):
        if first:
            node = self.root
        if node is None:
            return 0
        if node.left is None and node.right is None:
            return 1
        return self.count_leaves(node.left, False) + self.count_leaves(node.right, False)

    # 5, 6, 7, 8
    def count_degree(self):
        """Trả về (bậc 0, bậc 1, chỉ con phải, chỉ con trái, bậc 2)."""
        b0 = b1 = b2 = chi_phai = chi_trai = 0
        def fn(n):
            nonlocal b0, b1, b2, chi_phai, chi_trai
            con = (n.left is not None) + (n.right is not None)
            if con == 0:
                b0 += 1
            elif con == 1:
                b1 += 1
                if n.left is None: chi_phai += 1
                else: chi_trai += 1
            else:
                b2 += 1
        self._walk(self.root, fn)
        return b0, b1, chi_phai, chi_trai, b2

    # 9
    def print_by_level(self):
        if self.root is None:
            print("(cây rỗng)"); return []
        res, cur, k = [], [self.root], 0
        while any(cur):
            vals = [n.key for n in cur if n is not None]
            res.append(vals)
            print(f"   mức {k}: {vals}")
            nxt = []
            for n in cur:
                if n:
                    nxt.append(n.left); nxt.append(n.right)
            cur, k = nxt, k + 1
        return res

    # 10
    def path_length(self, x):
        """Độ dài đường đi từ gốc đến node x (số cạnh). -1 nếu không có x."""
        d, n = 0, self.root
        while n is not None:
            if x == n.key:
                return d
            n = n.left if x < n.key else n.right
            d += 1
        return -1

    def path_to(self, x):
        """Đường đi (danh sách giá trị) từ gốc đến x."""
        duong, n = [], self.root
        while n is not None:
            duong.append(n.key)
            if x == n.key:
                return duong
            n = n.left if x < n.key else n.right
        return None


ti = BSTInfo().build([7, 3, 36, 1, 6, 15, 40, 4, 23])
print("Cây (BT8):")
ti.draw()
print("1. Node chẵn:", end=' '); ti.print_even()
print("2. Số node            :", ti.count_nodes())
print("3. Độ cao             :", ti.height())
print("4. Số node lá         :", ti.count_leaves())
b0, b1, cp, ct, b2 = ti.count_degree()
print(f"5. Số node 1 cây con  : {b1}")
print(f"6. Chỉ có con phải    : {cp}")
print(f"7. Chỉ có con trái    : {ct}")
print(f"8. Số node 2 cây con  : {b2}")
print("9. Node theo từng mức :"); ti.print_by_level()
print("10. Độ dài đường đi gốc -> 23:", ti.path_length(23))
print("    Đường đi               :", ti.path_to(23))
```

```python
# ---- KIỂM THỬ BT8 ----
day_ti = [7, 3, 36, 1, 6, 15, 40, 4, 23]

# 1) node chẵn
assert sorted(ti.print_even()) == sorted(k for k in day_ti if k % 2 == 0)
print("Test 1 (node chẵn): PASS")

# 2) số node
assert ti.count_nodes() == len(day_ti)
print("Test 2 (số node == 9): PASS")

# 3) độ cao: đối chiếu với số mức
assert ti.height() == len(ti.levels())
print("Test 3 (độ cao == số mức): PASS")

# 4) số lá
def dem_la(n):
    if n is None: return 0
    if n.left is None and n.right is None: return 1
    return dem_la(n.left) + dem_la(n.right)
assert ti.count_leaves() == dem_la(ti.root)
print("Test 4 (số lá): PASS")

# 5-8) các loại bậc: tổng phải bằng số node
b0, b1, cp, ct, b2 = ti.count_degree()
assert b0 + b1 + b2 == ti.count_nodes(), "tổng bậc 0+1+2 phải bằng số node"
assert cp + ct == b1, "node 1 con = chỉ con phải + chỉ con trái"
assert b0 == b2 + 1, "định lý n0 = n2 + 1"
print(f"Test 5-8 (b0+b1+b2 = {b0}+{b1}+{b2} = số node; cp+ct = b1; n0 = n2+1): PASS")

# 9) in theo mức: tổng các mức = số node, và khớp với BFS
muc = ti.print_by_level()
assert sum(len(m) for m in muc) == ti.count_nodes()
assert [v for m in muc for v in m] == ti.breadth_first()
print("Test 9 (node theo mức == BFS): PASS")

# 10) độ dài đường đi: đối chiếu bằng cách duyệt tay
def do_dai_tay(n, x, d=0):
    if n is None: return -1
    if n.key == x: return d
    r = do_dai_tay(n.left, x, d + 1)
    return r if r != -1 else do_dai_tay(n.right, x, d + 1)
assert all(ti.path_length(x) == do_dai_tay(ti.root, x) for x in day_ti)
assert ti.path_length(999) == -1
print("Test 10 (độ dài đường đi, kể cả giá trị không tồn tại): PASS")

# đường đi phải bắt đầu ở gốc và kết thúc ở x
assert ti.path_to(23)[0] == ti.root.key and ti.path_to(23)[-1] == 23
assert ti.path_length(23) == len(ti.path_to(23)) - 1
print("Test path_to nhất quán với path_length: PASS")
```

## BT10 (tr. 55) — Xóa node giá trị x bằng 2 phương pháp

Viết hàm xóa node có giá trị `x` trong cây nhị phân số nguyên bằng hai phương pháp **Copying** và **Merging** (đã cài đặt ở BT2). Dưới đây kiểm thử **toàn diện**: với mọi nút của một cây, xóa bằng cả 2 phương pháp và so sánh.

```python
def kiem_tra_2_phuong_phap(keys):
    """Với mọi giá trị trong keys: xóa bằng cả 2 pp, so sánh tập giá trị và tính chất BST."""
    loi = []
    for x in keys:
        tA = BSTDelete().build(keys); tA.delete_copying(x)
        tB = BSTDelete().build(keys); tB.delete_merging(x)
        con_lai = sorted(k for k in keys if k != x)
        if tA.inorder() != con_lai:
            loi.append((x, "copying", tA.inorder()))
        if tB.inorder() != con_lai:
            loi.append((x, "merging", tB.inorder()))
        if not ktra(tA.root) or not ktra(tB.root):
            loi.append((x, "vi phạm BST", ""))
    return loi


import random

# 1) Trên cây của BT9
loi = kiem_tra_2_phuong_phap(day_bt9)
print("BT9 - xóa mọi nút bằng 2 phương pháp, số lỗi:", len(loi))
assert not loi

# 2) Trên nhiều cây ngẫu nhiên
random.seed(2024)
tong_loi = 0
for _ in range(30):
    keys = random.sample(range(1, 100), random.randint(3, 15))
    tong_loi += len(kiem_tra_2_phuong_phap(keys))
print("30 cây ngẫu nhiên - tổng số lỗi:", tong_loi)
assert tong_loi == 0
print("Test BT10 (2 phương pháp xóa, mọi nút, 30 cây ngẫu nhiên): PASS")

# 3) Xóa nút gốc và nút lá (trường hợp biên)
t = BSTDelete().build([50, 30, 70, 20, 40])
t.delete_copying(50)                      # gốc có 2 con
assert ktra(t.root) and t.inorder() == [20, 30, 40, 70]
t = BSTDelete().build([50, 30, 70])
t.delete_merging(20) if False else None
t.delete_merging(30)                      # nút lá
assert ktra(t.root) and t.inorder() == [50, 70]
print("Test BT10 trường hợp biên (xóa gốc, xóa lá): PASS")
```

## Phần mở rộng — Cây AVL (slide tr. 57–81)

Slide trình bày cây AVL với hệ số cân bằng $BF = h_{trái} - h_{phải} \in \{-1,0,1\}$ và 4 dạng mất cân bằng: **Left-Left** (xoay phải), **Right-Right** (xoay trái), **Left-Right** (xoay trái rồi xoay phải), **Right-Left** (xoay phải rồi xoay trái).

Dưới đây cài đặt đầy đủ AVL (chèn + xóa có cân bằng) và kiểm thử rằng cây **luôn cân bằng** sau mọi thao tác.

```python
class AVLNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.height = 1


class AVL:
    """Cây AVL: BST tự cân bằng (đúng theo code mẫu trong slide tr.80-81)."""

    def __init__(self):
        self.root = None

    def h(self, n):
        return n.height if n else 0

    def balance(self, n):
        return self.h(n.left) - self.h(n.right) if n else 0

    def update(self, n):
        n.height = 1 + max(self.h(n.left), self.h(n.right))

    # ---- 4 phép xoay ----
    def rotate_right(self, y):
        x = y.left
        t2 = x.right
        x.right = y
        y.left = t2
        self.update(y); self.update(x)
        return x

    def rotate_left(self, x):
        y = x.right
        t2 = y.left
        y.left = x
        x.right = t2
        self.update(x); self.update(y)
        return y

    # ---- Chèn ----
    def insert(self, key):
        self.root = self._insert(self.root, key)

    def _insert(self, node, key):
        if node is None:
            return AVLNode(key)
        if key < node.key:
            node.left = self._insert(node.left, key)
        elif key > node.key:
            node.right = self._insert(node.right, key)
        else:
            return node
        self.update(node)
        b = self.balance(node)
        # Left-Left
        if b > 1 and key < node.left.key:
            return self.rotate_right(node)
        # Right-Right
        if b < -1 and key > node.right.key:
            return self.rotate_left(node)
        # Left-Right
        if b > 1 and key > node.left.key:
            node.left = self.rotate_left(node.left)
            return self.rotate_right(node)
        # Right-Left
        if b < -1 and key < node.right.key:
            node.right = self.rotate_right(node.right)
            return self.rotate_left(node)
        return node

    # ---- Xóa ----
    def min_node(self, n):
        while n.left:
            n = n.left
        return n

    def delete(self, key):
        self.root = self._delete(self.root, key)

    def _delete(self, root, key):
        if root is None:
            return root
        if key < root.key:
            root.left = self._delete(root.left, key)
        elif key > root.key:
            root.right = self._delete(root.right, key)
        else:
            if root.left is None or root.right is None:
                root = root.left if root.left else root.right
            else:
                tmp = self.min_node(root.right)
                root.key = tmp.key
                root.right = self._delete(root.right, tmp.key)
        if root is None:
            return root
        self.update(root)
        b = self.balance(root)
        if b > 1 and self.balance(root.left) >= 0:
            return self.rotate_right(root)
        if b > 1 and self.balance(root.left) < 0:
            root.left = self.rotate_left(root.left)
            return self.rotate_right(root)
        if b < -1 and self.balance(root.right) <= 0:
            return self.rotate_left(root)
        if b < -1 and self.balance(root.right) > 0:
            root.right = self.rotate_right(root.right)
            return self.rotate_left(root)
        return root

    def inorder(self):
        out = []
        def w(n):
            if n:
                w(n.left); out.append(n.key); w(n.right)
        w(self.root)
        return out

    def height(self, n=None, first=True):
        if first: n = self.root
        return self.h(n)


# Kiểm thử: chèn 1..50 theo thứ tự tăng -> BST thường sẽ suy biến thành danh sách,
# AVL phải giữ chiều cao O(log n)
av = AVL()
for k in range(1, 51):
    av.insert(k)

print("Chèn 1..50 theo thứ tự tăng dần:")
print("  Số nút      :", len(av.inorder()))
print("  Chiều cao   :", av.height())
print("  log2(50) ≈ 5.6  ->  AVL giữ chiều cao O(log n), không suy biến")
print("  InOrder     :", av.inorder())
```

```python
# ---- KIỂM THỬ AVL ----
def ktra_can_bang(n):
    """Mọi nút phải có |BF| <= 1, và height ghi trong nút phải đúng."""
    if n is None:
        return True
    bf = (n.left.height if n.left else 0) - (n.right.height if n.right else 0)
    if abs(bf) > 1:
        return False
    h_dung = 1 + max(n.left.height if n.left else 0, n.right.height if n.right else 0)
    if n.height != h_dung:
        return False
    return ktra_can_bang(n.left) and ktra_can_bang(n.right)

assert ktra_can_bang(av.root), "AVL mất cân bằng sau khi chèn"
print("Test AVL cân bằng sau khi chèn 1..50 tăng dần: PASS")

# Chiều cao phải <= 1.44*log2(n+2) - 0.33 (chặn lý thuyết của AVL)
import math
n = len(av.inorder())
gioi_han = 1.4405 * math.log2(n + 2) - 0.3277
assert av.height() <= gioi_han, f"chiều cao {av.height()} vượt chặn lý thuyết {gioi_han:.2f}"
print(f"Test chiều cao {av.height()} <= chặn AVL {gioi_han:.2f}: PASS")

# Xóa dần toàn bộ, cây luôn cân bằng
import random
random.seed(1)
con = av.inorder()[:]
random.shuffle(con)
ok = True
for x in con:
    av.delete(x)
    if not ktra_can_bang(av.root):
        ok = False
        print(f"  mất cân bằng sau khi xóa {x}")
print("Test AVL luôn cân bằng khi xóa ngẫu nhiên toàn bộ 50 nút:", "PASS" if ok else "FAIL")
assert av.root is None
print("Test cây rỗng sau khi xóa hết: PASS")

# Chèn/xóa ngẫu nhiên: AVL luôn giữ tính chất BST + cân bằng
random.seed(99)
avl = AVL(); gia_tri = []
for _ in range(300):
    x = random.randint(1, 200)
    if random.random() < 0.6:
        avl.insert(x)
        if x not in gia_tri: gia_tri.append(x)
    elif gia_tri:
        y = random.choice(gia_tri); gia_tri.remove(y)
        avl.delete(y)
    assert ktra_can_bang(avl.root), "mất cân bằng"
    assert avl.inorder() == sorted(gia_tri), "sai tập giá trị"
print("Test 300 thao tác chèn/xóa ngẫu nhiên: AVL luôn là BST cân bằng: PASS")
```

## Phần mở rộng — Bảng băm (slide tr. 83–91)

Slide trình bày bảng băm với hàm băm, xử lý xung đột bằng **chaining** (danh sách liên kết), và các cách khác (open addressing, double hashing, linear/quadratic probing). Cài đặt lại và kiểm thử:

```python
class HashTable:
    """Bảng băm với chaining (đúng theo code mẫu trong slide tr.86-87)."""

    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]
        self.so_xung_dot = 0

    def _hash(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        i = self._hash(key)
        for pair in self.table[i]:
            if pair[0] == key:
                pair[1] = value          # cập nhật
                return
        if self.table[i]:
            self.so_xung_dot += 1        # xung đột: bucket đã có phần tử khác
        self.table[i].append([key, value])

    def get(self, key):
        i = self._hash(key)
        for pair in self.table[i]:
            if pair[0] == key:
                return pair[1]
        return None

    def delete(self, key):
        i = self._hash(key)
        for k, pair in enumerate(self.table[i]):
            if pair[0] == key:
                del self.table[i][k]
                return True
        return False

    def keys(self):
        return [p[0] for b in self.table for p in b]


ht = HashTable(10)
ht.insert("name", "Alice")
ht.insert("age", 30)
ht.insert("city", "Hanoi")
print("get('name') :", ht.get("name"))
print("get('age')  :", ht.get("age"))
ht.insert("age", 31)
print("sau khi cập nhật age ->", ht.get("age"))
ht.delete("city")
print("sau khi xóa city ->", ht.get("city"))
```

```python
# ---- KIỂM THỬ BẢNG BĂM ----
import random

# 1) Đối chiếu với dict của Python (vốn là bảng băm tối ưu)
random.seed(5)
ok = True
for _ in range(20):
    ht = HashTable(random.choice([5, 7, 11, 16]))
    d = {}
    for _ in range(200):
        k = random.randint(0, 60)
        v = random.randint(0, 1000)
        if random.random() < 0.75:
            ht.insert(k, v); d[k] = v
        else:
            ht.delete(k); d.pop(k, None)
        if ht.get(k) != d.get(k):
            ok = False
    if sorted(ht.keys()) != sorted(d.keys()):
        ok = False
print("Test HashTable == dict (chèn/xóa/truy vấn, 20 lượt x 200 thao tác):", "PASS" if ok else "FAIL")

# 2) Ví dụ trong slide
ht2 = HashTable(10)
ht2.insert("name", "Alice"); ht2.insert("age", 30); ht2.insert("city", "Hanoi")
assert ht2.get("name") == "Alice" and ht2.get("age") == 30
ht2.insert("age", 31)
assert ht2.get("age") == 31
ht2.delete("city")
assert ht2.get("city") is None
print("Test ví dụ trong slide: PASS")

# 3) Xử lý xung đột: khi nhồi nhiều khóa vào bảng nhỏ, vẫn tra cứu đúng
ht3 = HashTable(3)
for i in range(50):
    ht3.insert(i, i * i)
assert all(ht3.get(i) == i * i for i in range(50))
print(f"Test 50 khóa trong bảng size=3 (xảy ra {ht3.so_xung_dot} xung đột): PASS")
```

# Group Anagrams (Nhóm từ đảo)

## Bài toán
Cho một mảng các từ, hãy nhóm các từ đảo (anagram) lại với nhau.

> **Từ đảo (anagram):** hai từ gồm các ký tự giống nhau nhưng sắp xếp theo thứ tự khác.
> Ví dụ: `"eat"`, `"tea"`, `"ate"` là các từ đảo của nhau.

## Ý tưởng thuật toán
1. Với mỗi từ, **sắp xếp các ký tự** theo thứ tự alphabet để tạo một **key chuẩn hóa**.
2. Dùng **hash map (dictionary)** để nhóm các từ có cùng key lại với nhau.
3. Trả về danh sách các nhóm.

## Độ phức tạp
| | Thời gian | Không gian |
|---|---|---|
| **Phân tích** | O(n · k log k) | O(n · k) |

- `n` = số từ, `k` = độ dài từ dài nhất
- Mỗi từ cần sort k ký tự → O(k log k)


````python
# nhóm từ đảo (group anagrams)
from collections import defaultdict
def group_anagrams(words):
    groups = defaultdict(list)
    for word in words:
        key = tuple(sorted(word))
        groups[key].append(word)
    return list(groups.values())
print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))

    
    
````

## Giải thích code

```python
from collections import defaultdict

def group_anagrams(words):
    groups = defaultdict(list)      # tạo dict với mặc định value là list rỗng
    for word in words:
        key = tuple(sorted(word))    # sort ký tự → key chuẩn hóa
        groups[key].append(word)     # nhóm từ vào đúng key
    return list(groups.values())     # trả về các nhóm
```

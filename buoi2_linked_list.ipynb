## BT1 (tr. 24) — Quản lý danh sách sinh viên bằng `list`

Lập trình menu quản lý danh sách N sinh viên với các trường `{Mã sinh viên, Họ tên, Điểm GPA}`:
1. Nhập N sinh viên
2. In danh sách sinh viên
3. Thêm một sinh viên vào vị trí `pos`
4. Tìm một sinh viên theo mã sinh viên
5. Xóa một sinh viên theo mã sinh viên
6. Sắp xếp danh sách sinh viên theo điểm GPA tăng dần

> Để notebook chạy được không cần nhập tay, dữ liệu mẫu được nạp sẵn; menu tương tác vẫn được cài đặt đầy đủ ở cuối mục.

```python
class SinhVien:
    def __init__(self, ma, ho_ten, gpa):
        self.ma = ma
        self.ho_ten = ho_ten
        self.gpa = gpa

    def __str__(self):
        return f"{self.ma:<8} {self.ho_ten:<22} {self.gpa:>5.2f}"

    def __repr__(self):
        return f"SinhVien({self.ma!r}, {self.ho_ten!r}, {self.gpa})"


class QuanLySV:
    """Quản lý danh sách sinh viên bằng list (BT1)."""

    def __init__(self):
        self.ds = []

    # 1. Nhập N sinh viên
    def nhap(self, ds_input):
        for ma, ten, gpa in ds_input:
            self.ds.append(SinhVien(ma, ten, gpa))
        return len(self.ds)

    # 2. In danh sách
    def in_ds(self):
        if not self.ds:
            print("(danh sách rỗng)")
            return
        print(f"{'Mã SV':<8} {'Họ tên':<22} {'GPA':>5}")
        print("-" * 37)
        for sv in self.ds:
            print(sv)

    # 3. Thêm sinh viên vào vị trí pos
    def them_tai(self, pos, sv):
        if pos < 0 or pos > len(self.ds):
            return False
        self.ds.insert(pos, sv)
        return True

    # 4. Tìm theo mã sinh viên
    def tim(self, ma):
        for sv in self.ds:
            if sv.ma == ma:
                return sv
        return None

    # 5. Xóa theo mã sinh viên
    def xoa(self, ma):
        for i, sv in enumerate(self.ds):
            if sv.ma == ma:
                del self.ds[i]
                return True
        return False

    # 6. Sắp xếp theo GPA tăng dần
    def sap_xep_gpa(self):
        self.ds.sort(key=lambda sv: sv.gpa)


# ---- Dữ liệu mẫu ----
du_lieu = [
    ("SV01", "Nguyễn Văn An",   7.5),
    ("SV02", "Trần Thị Bình",   8.9),
    ("SV03", "Lê Hoàng Cường",  6.4),
    ("SV04", "Phạm Thu Dung",   9.2),
    ("SV05", "Võ Minh Đức",     5.8),
]

ql = QuanLySV()
print("1. Nhập", ql.nhap(du_lieu), "sinh viên.\n")
print("2. Danh sách sinh viên:")
ql.in_ds()

print("\n3. Thêm SV06 vào vị trí pos = 2:")
ql.them_tai(2, SinhVien("SV06", "Đỗ Khánh Hà", 7.9))
ql.in_ds()

print("\n4. Tìm sinh viên mã SV03:", ql.tim("SV03"))
print("   Tìm sinh viên mã SV99:", ql.tim("SV99"))

print("\n5. Xóa sinh viên mã SV05:", ql.xoa("SV05"))
print("\n6. Sắp xếp theo GPA tăng dần:")
ql.sap_xep_gpa()
ql.in_ds()
```

```python
# ---- KIỂM THỬ BT1 ----
ql2 = QuanLySV()
ql2.nhap(du_lieu)

# 1) Số lượng sau khi nhập
assert len(ql2.ds) == 5
print("Test nhập 5 sinh viên: PASS")

# 2) Thêm tại vị trí: đúng vị trí, đúng thứ tự
ql2.them_tai(2, SinhVien("SV06", "Đỗ Khánh Hà", 7.9))
assert [sv.ma for sv in ql2.ds] == ["SV01", "SV02", "SV06", "SV03", "SV04", "SV05"]
assert ql2.them_tai(99, SinhVien("SV99", "x", 1.0)) is False     # pos ngoài phạm vi
print("Test thêm tại pos (kể cả pos không hợp lệ): PASS")

# 3) Tìm theo mã
assert ql2.tim("SV03").ho_ten == "Lê Hoàng Cường"
assert ql2.tim("SV99") is None
print("Test tìm theo mã (có / không có): PASS")

# 4) Xóa theo mã
assert ql2.xoa("SV05") is True
assert ql2.xoa("SV05") is False                                  # xóa lần 2 phải thất bại
assert "SV05" not in [sv.ma for sv in ql2.ds]
print("Test xóa theo mã (xóa lần 2 trả False): PASS")

# 5) Sắp xếp theo GPA tăng dần: kiểm tra dãy không giảm
ql2.sap_xep_gpa()
gpas = [sv.gpa for sv in ql2.ds]
assert gpas == sorted(gpas), gpas
print("Test sắp xếp GPA tăng dần:", gpas, "-> PASS")

# 6) Đối chiếu với sorted() của Python
assert [sv.ma for sv in ql2.ds] == [sv.ma for sv in sorted(ql2.ds, key=lambda s: s.gpa)]
print("Test đối chiếu với sorted(): PASS")
```

## BT2 (tr. 37) — Quản lý N số nguyên bằng DSLK đơn tự xây

Menu 10 thao tác: in danh sách; thêm đầu; thêm cuối; thêm tại `pos`; xóa đầu; xóa cuối; xóa tại `pos`; xóa giá trị `X`; tìm `X`; sắp xếp.

```python
class Node:
    """Một nút của danh sách liên kết đơn (theo slide tr.26)."""
    def __init__(self, info):
        self.info = info
        self.next = None


class DSLKDon:
    """Danh sách liên kết đơn tự xây dựng."""

    def __init__(self):
        self.head = None
        self.n = 0

    # 1. In danh sách
    def in_ds(self):
        out, p = [], self.head
        while p:
            out.append(p.info)
            p = p.next
        print("Danh sách:", out)
        return out

    def to_list(self):
        out, p = [], self.head
        while p:
            out.append(p.info)
            p = p.next
        return out

    # 2. Thêm đầu
    def them_dau(self, x):
        p = Node(x)
        p.next = self.head
        self.head = p
        self.n += 1

    # 3. Thêm cuối
    def them_cuoi(self, x):
        p = Node(x)
        if self.head is None:
            self.head = p
        else:
            q = self.head
            while q.next:
                q = q.next
            q.next = p
        self.n += 1

    # 4. Thêm tại vị trí pos (0-based)
    def them_tai(self, pos, x):
        if pos < 0 or pos > self.n:
            return False
        if pos == 0:
            self.them_dau(x)
            return True
        p = Node(x)
        q = self.head
        for _ in range(pos - 1):
            q = q.next
        p.next = q.next
        q.next = p
        self.n += 1
        return True

    # 5. Xóa đầu
    def xoa_dau(self):
        if self.head is None:
            return False
        self.head = self.head.next
        self.n -= 1
        return True

    # 6. Xóa cuối
    def xoa_cuoi(self):
        if self.head is None:
            return False
        if self.head.next is None:
            self.head = None
        else:
            q = self.head
            while q.next.next:
                q = q.next
            q.next = None
        self.n -= 1
        return True

    # 7. Xóa tại vị trí pos
    def xoa_tai(self, pos):
        if pos < 0 or pos >= self.n:
            return False
        if pos == 0:
            return self.xoa_dau()
        q = self.head
        for _ in range(pos - 1):
            q = q.next
        q.next = q.next.next
        self.n -= 1
        return True

    # 8. Xóa phần tử có giá trị X (đầu tiên)
    def xoa_gia_tri(self, x):
        if self.head is None:
            return False
        if self.head.info == x:
            return self.xoa_dau()
        q = self.head
        while q.next and q.next.info != x:
            q = q.next
        if q.next is None:
            return False
        q.next = q.next.next
        self.n -= 1
        return True

    # 9. Tìm X -> trả về vị trí (hoặc -1)
    def tim(self, x):
        p, i = self.head, 0
        while p:
            if p.info == x:
                return i
            p, i = p.next, i + 1
        return -1

    # 10. Sắp xếp (selection sort trên DSLK)
    def sap_xep(self):
        if self.head is None:
            return
        p = self.head
        while p:
            q = p.next
            while q:
                if q.info < p.info:
                    p.info, q.info = q.info, p.info
                q = q.next
            p = p.next


ds = DSLKDon()
for x in [5, 3, 8, 1, 9]:
    ds.them_cuoi(x)
print("Khởi tạo bằng thêm cuối [5,3,8,1,9]:")
ds.in_ds()

print("\nThêm đầu 0:"); ds.them_dau(0); ds.in_ds()
print("Thêm cuối 7:"); ds.them_cuoi(7); ds.in_ds()
print("Thêm 99 tại pos=3:"); ds.them_tai(3, 99); ds.in_ds()
print("Xóa đầu:"); ds.xoa_dau(); ds.in_ds()
print("Xóa cuối:"); ds.xoa_cuoi(); ds.in_ds()
print("Xóa tại pos=2:"); ds.xoa_tai(2); ds.in_ds()
print("Xóa giá trị 8:"); ds.xoa_gia_tri(8); ds.in_ds()
print("Tìm 9 -> vị trí", ds.tim(9), "| Tìm 100 ->", ds.tim(100))
print("Sắp xếp:"); ds.sap_xep(); ds.in_ds()
```

## BT3 (tr. 42) — 4 ý

1. Nhập dãy số nguyên → DSLK đơn, xuất dãy, cài đặt thêm/xóa
2. Quản lý danh sách sinh viên bằng **DSLKĐ**: tạo n SV, xuất, xuất SV có ĐTB > 5, tìm SV tên X, sắp tăng theo ĐTB
3. Hiện thực **danh sách liên kết đôi**
4. Hiện thực **danh sách liên kết vòng**

```python
# ---------- Ý 3: DANH SÁCH LIÊN KẾT ĐÔI ----------
class Node2:
    def __init__(self, info):
        self.info = info
        self.prev = None
        self.next = None


class DSLKDoi:
    """Danh sách liên kết đôi (slide tr.40-41): duyệt được 2 chiều."""

    def __init__(self):
        self.head = None
        self.tail = None
        self.n = 0

    def to_list(self):
        out, p = [], self.head
        while p:
            out.append(p.info)
            p = p.next
        return out

    def to_list_nguoc(self):
        out, p = [], self.tail
        while p:
            out.append(p.info)
            p = p.prev
        return out

    def them_dau(self, x):
        p = Node2(x)
        p.next = self.head
        if self.head:
            self.head.prev = p
        else:
            self.tail = p
        self.head = p
        self.n += 1

    def them_cuoi(self, x):
        p = Node2(x)
        p.prev = self.tail
        if self.tail:
            self.tail.next = p
        else:
            self.head = p
        self.tail = p
        self.n += 1

    def xoa(self, x):
        """Xóa phần tử đầu tiên có giá trị x."""
        p = self.head
        while p and p.info != x:
            p = p.next
        if p is None:
            return False
        if p.prev:
            p.prev.next = p.next
        else:
            self.head = p.next
        if p.next:
            p.next.prev = p.prev
        else:
            self.tail = p.prev
        self.n -= 1
        return True

    def in_ds(self):
        print("Xuôi :", self.to_list())
        print("Ngược:", self.to_list_nguoc())


# ---------- Ý 4: DANH SÁCH LIÊN KẾT VÒNG ----------
class DSLKVong:
    """Danh sách liên kết vòng (slide tr.38-39): next của nút cuối trỏ về đầu."""

    def __init__(self):
        self.tail = None       # theo slide: pList trỏ đến phần tử CUỐI
        self.n = 0

    def them_cuoi(self, x):
        p = Node(x)
        if self.tail is None:
            p.next = p            # tự trỏ về chính nó
            self.tail = p
        else:
            p.next = self.tail.next   # nút đầu
            self.tail.next = p
            self.tail = p
        self.n += 1

    def them_dau(self, x):
        p = Node(x)
        if self.tail is None:
            p.next = p
            self.tail = p
        else:
            p.next = self.tail.next
            self.tail.next = p
        self.n += 1

    def xoa_dau(self):
        if self.tail is None:
            return False
        if self.tail.next is self.tail:      # chỉ có 1 nút
            self.tail = None
        else:
            self.tail.next = self.tail.next.next
        self.n -= 1
        return True

    def to_list(self, gioi_han=None):
        """Duyệt vòng; cần giới hạn số bước để không lặp vô hạn."""
        if self.tail is None:
            return []
        out, p, dem = [], self.tail.next, 0
        limit = gioi_han if gioi_han is not None else self.n
        while dem < limit:
            out.append(p.info)
            p = p.next
            dem += 1
        return out

    def in_ds(self):
        print("DSLK vòng:", self.to_list())


print("=== Ý 3: DSLK đôi ===")
d2 = DSLKDoi()
for x in [3, 1, 4, 1, 5]:
    d2.them_cuoi(x)
d2.them_dau(0)
d2.in_ds()
print("Xóa 1:", d2.xoa(1)); d2.in_ds()

print("\n=== Ý 4: DSLK vòng ===")
dv = DSLKVong()
for x in [1, 2, 3, 4]:
    dv.them_cuoi(x)
dv.in_ds()
dv.them_dau(0)
dv.in_ds()
dv.xoa_dau()
dv.in_ds()
```

## BT4 (tr. 43) — Quản lý N số nguyên bằng DSLK đôi

Menu 10 thao tác, trong đó chú ý:
- ý 7: xóa phần tử có giá trị `X` **đầu tiên**
- ý 8: xóa **tất cả** phần tử có giá trị `X`

```python
class DSLKDoiFull(DSLKDoi):
    """Bổ sung các thao tác của BT4."""

    def them_tai(self, pos, x):
        if pos < 0 or pos > self.n:
            return False
        if pos == 0:
            self.them_dau(x); return True
        if pos == self.n:
            self.them_cuoi(x); return True
        p = self.head
        for _ in range(pos):
            p = p.next
        q = Node2(x)
        q.prev = p.prev
        q.next = p
        p.prev.next = q
        p.prev = q
        self.n += 1
        return True

    def xoa_dau(self):
        if self.head is None:
            return False
        self.head = self.head.next
        if self.head:
            self.head.prev = None
        else:
            self.tail = None
        self.n -= 1
        return True

    def xoa_cuoi(self):
        if self.tail is None:
            return False
        self.tail = self.tail.prev
        if self.tail:
            self.tail.next = None
        else:
            self.head = None
        self.n -= 1
        return True

    def xoa_tai(self, pos):
        if pos < 0 or pos >= self.n:
            return False
        if pos == 0:
            return self.xoa_dau()
        if pos == self.n - 1:
            return self.xoa_cuoi()
        p = self.head
        for _ in range(pos):
            p = p.next
        p.prev.next = p.next
        p.next.prev = p.prev
        self.n -= 1
        return True

    def xoa_dau_tien(self, x):
        """ý 7: xóa phần tử ĐẦU TIÊN có giá trị x."""
        return self.xoa(x)

    def xoa_tat_ca(self, x):
        """ý 8: xóa TẤT CẢ phần tử có giá trị x."""
        dem = 0
        while self.xoa(x):
            dem += 1
        return dem

    def tim_dau_tien(self, x):
        """ý 9: tìm vị trí phần tử đầu tiên có giá trị x."""
        p, i = self.head, 0
        while p:
            if p.info == x:
                return i
            p, i = p.next, i + 1
        return -1

    def sap_xep(self):
        """ý 10: sắp xếp tăng dần."""
        p = self.head
        while p:
            q = p.next
            while q:
                if q.info < p.info:
                    p.info, q.info = q.info, p.info
                q = q.next
            p = p.next


d4 = DSLKDoiFull()
for x in [4, 2, 7, 2, 9, 2, 1]:
    d4.them_cuoi(x)
print("Khởi tạo:", d4.to_list())
print("Thêm 5 tại pos=3:"); d4.them_tai(3, 5); print("  ", d4.to_list())
print("Xóa đầu:"); d4.xoa_dau(); print("  ", d4.to_list())
print("Xóa cuối:"); d4.xoa_cuoi(); print("  ", d4.to_list())
print("Xóa tại pos=2:"); d4.xoa_tai(2); print("  ", d4.to_list())
print("Xóa phần tử ĐẦU TIÊN có giá trị 2:"); d4.xoa_dau_tien(2); print("  ", d4.to_list())
print("Xóa TẤT CẢ giá trị 2:"); print("   đã xóa", d4.xoa_tat_ca(2), "phần tử ->", d4.to_list())
print("Tìm 9 -> vị trí", d4.tim_dau_tien(9))
d4.sap_xep(); print("Sắp xếp:", d4.to_list())
```

## BT5 (tr. 60) — Chuyển infix → postfix và tính giá trị

Biểu thức: `10 * 2 + (5*(8 + 4)) - 3`

**Thuật toán chuyển Infix → Postfix** (slide tr.57):
1. Khởi tạo `opStack` rỗng và `list` kết quả rỗng
2. Tách chuỗi thành các token
3. Duyệt token trái → phải:
   - toán hạng → thêm vào cuối `list`
   - `(` → push vào `opStack`
   - `)` → pop `opStack` cho đến khi gặp `(`
   - toán tử → trong khi đỉnh `opStack` là toán tử có độ ưu tiên ≥ độ ưu tiên của token thì pop; rồi push token
4. Hết token → pop hết `opStack` còn lại vào `list`

**Thuật toán tính Postfix** (slide tr.59): gặp toán hạng thì push; gặp toán tử thì `b = pop`, `a = pop`, push `a op b`.

```python
UU_TIEN = {'+': 1, '-': 1, '*': 2, '/': 2}


def tach_token(bieu_thuc):
    """Tách biểu thức thành token (số nhiều chữ số, toán tử, ngoặc)."""
    toks, i, n = [], 0, len(bieu_thuc)
    while i < n:
        c = bieu_thuc[i]
        if c.isspace():
            i += 1
        elif c.isdigit():
            j = i
            while j < n and bieu_thuc[j].isdigit():
                j += 1
            toks.append(bieu_thuc[i:j])
            i = j
        else:
            toks.append(c)
            i += 1
    return toks


def infix_to_postfix(bieu_thuc, verbose=False):
    """Chuyển infix -> postfix dùng stack (đúng thuật toán trong slide)."""
    op_stack = []
    ket_qua = []
    for tok in tach_token(bieu_thuc):
        if verbose:
            print(f"   token {tok!r:>6} | stack={op_stack} | kq={ket_qua}")
        if tok.isdigit():
            ket_qua.append(tok)
        elif tok == '(':
            op_stack.append(tok)
        elif tok == ')':
            while op_stack and op_stack[-1] != '(':
                ket_qua.append(op_stack.pop())
            op_stack.pop()                       # bỏ dấu (
        else:                                    # toán tử
            while (op_stack and op_stack[-1] != '(' and
                   UU_TIEN.get(op_stack[-1], 0) >= UU_TIEN[tok]):
                ket_qua.append(op_stack.pop())
            op_stack.append(tok)
    while op_stack:
        ket_qua.append(op_stack.pop())
    return ket_qua


def tinh_postfix(postfix, verbose=False):
    """Tính giá trị biểu thức postfix dùng stack."""
    s = []
    for tok in postfix:
        if tok.isdigit():
            s.append(int(tok))
        else:
            b = s.pop()
            a = s.pop()
            if tok == '+': kq = a + b
            elif tok == '-': kq = a - b
            elif tok == '*': kq = a * b
            else: kq = a / b
            s.append(kq)
        if verbose:
            print(f"   token {tok!r:>6} | stack={s}")
    return s.pop()


bt5 = "10 * 2 + (5*(8 + 4)) - 3"
print("Biểu thức:", bt5)
print("\n--- Chuyển infix -> postfix (từng bước) ---")
pf5 = infix_to_postfix(bt5, verbose=True)
print("\nPostfix:", " ".join(pf5))
print("\n--- Tính giá trị postfix (từng bước) ---")
kq5 = tinh_postfix(pf5, verbose=True)
print("\nKết quả =", kq5)
print("Kiểm tra bằng Python:", eval(bt5))
```

## BT6 (tr. 61) — Minh họa từng bước với stack

Biểu thức: `(6+7) * 2 + ((2+3)*(8 + 4))`

Yêu cầu: **1)** chuyển infix → postfix từng bước; **2)** tính giá trị postfix từng bước.

```python
bt6 = "(6+7) * 2 + ((2+3)*(8 + 4))"
print("Biểu thức:", bt6)
print("\n========== BƯỚC 1: INFIX -> POSTFIX (từng bước) ==========")
print(f"{'Token':>8} | {'opStack':<28} | Kết quả")
print("-" * 70)

op_stack, ket_qua = [], []
for tok in tach_token(bt6):
    if tok.isdigit():
        ket_qua.append(tok)
    elif tok == '(':
        op_stack.append(tok)
    elif tok == ')':
        while op_stack and op_stack[-1] != '(':
            ket_qua.append(op_stack.pop())
        op_stack.pop()
    else:
        while (op_stack and op_stack[-1] != '(' and
               UU_TIEN.get(op_stack[-1], 0) >= UU_TIEN[tok]):
            ket_qua.append(op_stack.pop())
        op_stack.append(tok)
    print(f"{tok:>8} | {str(op_stack):<28} | {ket_qua}")
while op_stack:
    ket_qua.append(op_stack.pop())
    print(f"{'(hết)':>8} | {str(op_stack):<28} | {ket_qua}")

print("\n=> Postfix:", " ".join(ket_qua))
print("\n========== BƯỚC 2: TÍNH GIÁ TRỊ POSTFIX (từng bước) ==========")
print(f"{'Token':>8} | Stack")
print("-" * 50)
s = []
for tok in ket_qua:
    if tok.isdigit():
        s.append(int(tok))
    else:
        b, a = s.pop(), s.pop()
        kq = a + b if tok == '+' else a - b if tok == '-' else a * b if tok == '*' else a / b
        s.append(kq)
    print(f"{tok:>8} | {s}")
print("\n=> Kết quả =", s[0])
print("Kiểm tra bằng Python:", eval(bt6))
```

## BT7 (tr. 71) — Stack bằng DSLK để tính biểu thức

Sử dụng **Stack cài bằng danh sách liên kết** để tính giá trị biểu thức nhập vào, ví dụ `3+6*(2+3) = 33`.

Hướng dẫn của đề: 1) cài đặt stack bằng DSLK; 2) chuyển infix → postfix; 3) dùng stack tính postfix.

```python
class StackDSLK:
    """Stack cài đặt bằng danh sách liên kết đơn (LIFO)."""

    def __init__(self):
        self.head = None
        self.n = 0

    def is_empty(self):
        return self.head is None

    def push(self, x):
        p = Node(x)
        p.next = self.head
        self.head = p
        self.n += 1

    def pop(self):
        if self.head is None:
            return None
        x = self.head.info
        self.head = self.head.next
        self.n -= 1
        return x

    def top(self):
        return None if self.head is None else self.head.info

    def __len__(self):
        return self.n


def infix_to_postfix_stackdslk(bieu_thuc):
    """Chuyển infix -> postfix, dùng Stack cài bằng DSLK."""
    op_stack = StackDSLK()
    ket_qua = []
    for tok in tach_token(bieu_thuc):
        if tok.isdigit():
            ket_qua.append(tok)
        elif tok == '(':
            op_stack.push(tok)
        elif tok == ')':
            while not op_stack.is_empty() and op_stack.top() != '(':
                ket_qua.append(op_stack.pop())
            op_stack.pop()
        else:
            while (not op_stack.is_empty() and op_stack.top() != '(' and
                   UU_TIEN.get(op_stack.top(), 0) >= UU_TIEN[tok]):
                ket_qua.append(op_stack.pop())
            op_stack.push(tok)
    while not op_stack.is_empty():
        ket_qua.append(op_stack.pop())
    return ket_qua


def tinh_postfix_stackdslk(postfix):
    """Tính postfix, dùng Stack cài bằng DSLK."""
    s = StackDSLK()
    for tok in postfix:
        if tok.isdigit():
            s.push(int(tok))
        else:
            b = s.pop(); a = s.pop()
            if tok == '+': s.push(a + b)
            elif tok == '-': s.push(a - b)
            elif tok == '*': s.push(a * b)
            else: s.push(a / b)
    return s.pop()


bt7 = "3+6*(2+3)"
print("Biểu thức:", bt7)
pf7 = infix_to_postfix_stackdslk(bt7)
print("Postfix  :", " ".join(pf7))
print("Kết quả  :", tinh_postfix_stackdslk(pf7), " (đề cho biết = 33)")
```

```python
# ---- KIỂM THỬ BT7 ----
# 1) Ví dụ của đề: 3+6*(2+3) = 33
assert tinh_postfix_stackdslk(infix_to_postfix_stackdslk("3+6*(2+3)")) == 33
print("Test BT7 ví dụ đề bài (3+6*(2+3) = 33): PASS")

# 2) Stack DSLK phải hoạt động đúng như stack list
random.seed(4)
s1, s2 = StackDSLK(), []
ok = True
for _ in range(500):
    if random.random() < 0.6 or not s2:
        x = random.randint(0, 99)
        s1.push(x); s2.append(x)
    else:
        if s1.pop() != s2.pop():
            ok = False; break
    if len(s1) != len(s2):
        ok = False; break
print("Test Stack DSLK == stack list (500 thao tác):", "PASS" if ok else "FAIL")

# 3) Stack DSLK == Stack list trên nhiều biểu thức
def infix_to_postfix_list(bt):
    op, kq = [], []
    for tok in tach_token(bt):
        if tok.isdigit(): kq.append(tok)
        elif tok == '(': op.append(tok)
        elif tok == ')':
            while op and op[-1] != '(': kq.append(op.pop())
            op.pop()
        else:
            while op and op[-1] != '(' and UU_TIEN.get(op[-1], 0) >= UU_TIEN[tok]:
                kq.append(op.pop())
            op.append(tok)
    while op: kq.append(op.pop())
    return kq

random.seed(12)
ok = True
for _ in range(200):
    bt = sinh_bieu_thuc()
    if infix_to_postfix_stackdslk(bt) != infix_to_postfix_list(bt):
        ok = False; print("SAI:", bt); break
    if tinh_postfix_stackdslk(infix_to_postfix_stackdslk(bt)) != eval(bt):
        ok = False; print("SAI giá trị:", bt); break
print("Test 200 biểu thức: Stack DSLK == Stack list == eval:", "PASS" if ok else "FAIL")

# 4) Trường hợp biên: stack rỗng
s = StackDSLK()
assert s.is_empty() and s.pop() is None and s.top() is None and len(s) == 0
print("Test Stack DSLK rỗng (pop/top trả None): PASS")
```

## BT8 (tr. 72) & BT9 (tr. 73) — Sinh ngẫu nhiên, Stack bằng List / DSLK

**BT8:** sinh ngẫu nhiên n số nguyên, đẩy vào Stack cài bằng **List**; in các phần tử theo thứ tự **đẩy vào**; rồi lấy ra và in theo thứ tự **đẩy ra**.

**BT9:** như trên nhưng Stack cài bằng **danh sách liên kết**.

> Vì Stack là LIFO, thứ tự đẩy ra là **ngược** với thứ tự đẩy vào.

```python
import random

class StackList:
    """Stack cài bằng list (dùng append/pop ở cuối)."""

    def __init__(self):
        self.a = []

    def is_empty(self):
        return len(self.a) == 0

    def push(self, x):
        self.a.append(x)

    def pop(self):
        return None if not self.a else self.a.pop()

    def top(self):
        return None if not self.a else self.a[-1]

    def __len__(self):
        return len(self.a)


def bai_8_9(n, loai="list", seed=42):
    """Sinh n số ngẫu nhiên, đẩy vào Stack, in thứ tự đẩy vào / đẩy ra."""
    random.seed(seed)
    day = [random.randint(1, 99) for _ in range(n)]
    st = StackList() if loai == "list" else StackDSLK()

    for x in day:
        st.push(x)
    print(f"  Đã đẩy vào Stack ({loai}): {day}")

    ra = []
    while not st.is_empty():
        ra.append(st.pop())
    print(f"  Thứ tự đẩy RA           : {ra}")
    return day, ra


print("BT8 - Stack cài bằng LIST (n = 10):")
day8, ra8 = bai_8_9(10, "list")

print("\nBT9 - Stack cài bằng DSLK (n = 10):")
day9, ra9 = bai_8_9(10, "dslk")
```

## BT10 (tr. 74) — Stack mảng 1 chiều & Stack DSLK; nhập số nguyên dương; copy Stack

1. Viết hàm nhập, xuất Stack số nguyên dương: nhập lần lượt, **dừng khi giá trị ≤ 0**
2. Viết hàm **copy Stack**

```python
class StackMang:
    """Stack cài bằng mảng 1 chiều với kích thước tối đa."""

    def __init__(self, max_size=100):
        self.a = [None] * max_size
        self.top_idx = -1
        self.max_size = max_size

    def is_empty(self):
        return self.top_idx == -1

    def is_full(self):
        return self.top_idx == self.max_size - 1

    def push(self, x):
        if self.is_full():
            return False
        self.top_idx += 1
        self.a[self.top_idx] = x
        return True

    def pop(self):
        if self.is_empty():
            return None
        x = self.a[self.top_idx]
        self.top_idx -= 1
        return x

    def top(self):
        return None if self.is_empty() else self.a[self.top_idx]

    def __len__(self):
        return self.top_idx + 1

    def to_list(self):
        """Từ đáy đến đỉnh."""
        return self.a[:self.top_idx + 1]


def nhap_stack(ds_input, st):
    """Nhập lần lượt vào Stack, DỪNG khi gặp giá trị <= 0."""
    for x in ds_input:
        if x <= 0:
            print(f"  Gặp giá trị {x} <= 0 -> dừng nhập.")
            break
        st.push(x)
    return st


def xuat_stack(st):
    """Xuất Stack (từ đáy đến đỉnh)."""
    print("  Stack (đáy -> đỉnh):", st.to_list())


def copy_stack(st):
    """Copy Stack: dùng 1 stack phụ để đảo lại đúng thứ tự."""
    tam = StackMang() if isinstance(st, StackMang) else StackDSLK()
    while not st.is_empty():
        tam.push(st.pop())
    moi = StackMang() if isinstance(st, StackMang) else StackDSLK()
    while not tam.is_empty():
        x = tam.pop()
        st.push(x)      # phục hồi stack gốc
        moi.push(x)     # stack mới
    return moi


print("=== Stack MẢNG 1 CHIỀU ===")
sm = StackMang()
nhap_stack([5, 8, 3, 12, 7, 0, 99], sm)      # dừng ở 0
xuat_stack(sm)
sm_copy = copy_stack(sm)
print("  Bản copy           :", sm_copy.to_list())
print("  Stack gốc nguyên vẹn:", sm.to_list())

print("\n=== Stack DSLK ===")
sd = StackDSLK()
nhap_stack([4, 9, 2, -1, 100], sd)            # dừng ở -1
print("  Stack (đáy -> đỉnh):", sd.to_list() if hasattr(sd, 'to_list') else [])
```

```python
# ---- KIỂM THỬ BT10 ----
# Bổ sung to_list cho StackDSLK để so sánh
def dslk_to_list(st):
    out, p = [], st.head
    while p:
        out.append(p.info)
        p = p.next
    return out[::-1]        # từ đáy đến đỉnh


# 1) Dừng khi gặp giá trị <= 0
sm = StackMang()
nhap_stack([5, 8, 3, 12, 7, 0, 99], sm)
assert sm.to_list() == [5, 8, 3, 12, 7], sm.to_list()
print("Test dừng nhập khi gặp 0 (bỏ cả 99 phía sau): PASS")

sd = StackDSLK()
nhap_stack([4, 9, 2, -1, 100], sd)
assert dslk_to_list(sd) == [4, 9, 2]
print("Test dừng nhập khi gặp -1: PASS")

# 2) Copy Stack: giữ nguyên stack gốc và bản copy giống hệt
for st, getter in [(StackMang(), lambda s: s.to_list()), (StackDSLK(), dslk_to_list)]:
    nhap_stack([1, 2, 3, 4, 5], st)
    truoc = getter(st)
    cp = copy_stack(st)
    assert getter(st) == truoc, "stack gốc bị thay đổi!"
    assert getter(cp) == truoc, "bản copy khác stack gốc!"
print("Test copy Stack (cả mảng và DSLK): gốc nguyên vẹn, copy giống hệt: PASS")

# 3) Copy rồi pop bản copy không ảnh hưởng gốc
sm = StackMang(); nhap_stack([1, 2, 3], sm)
cp = copy_stack(sm)
cp.pop()
assert sm.to_list() == [1, 2, 3] and cp.to_list() == [1, 2]
print("Test pop trên bản copy không ảnh hưởng stack gốc: PASS")

# 4) Stack mảng: đầy thì push trả False
sm = StackMang(max_size=3)
assert sm.push(1) and sm.push(2) and sm.push(3)
assert sm.is_full() and sm.push(4) is False
assert sm.to_list() == [1, 2, 3]
print("Test Stack mảng đầy (push thứ 4 trả False): PASS")

# 5) Copy stack rỗng
sm = StackMang()
cp = copy_stack(sm)
assert sm.is_empty() and cp.is_empty()
print("Test copy Stack rỗng: PASS")
```

## BT11 (tr. 75) — Stack từ chuỗi; QuickSort dùng Stack; Infix → Postfix

1. Tạo Stack từ chuỗi ký tự `str` (mỗi phần tử là 1 ký tự)
2. Tạo Stack từ chuỗi ký tự `str` (mỗi phần tử là 1 **từ**, cách nhau bởi khoảng trắng)
3. Cài đặt **QuickSort dùng Stack** (khử đệ quy)
4. Cài đặt chương trình chuyển **Infix → Postfix**

```python
# ---------- 1 & 2: Stack từ chuỗi ----------
def stack_tu_ky_tu(s):
    """Mỗi phần tử Stack là một KÝ TỰ."""
    st = StackList()
    for ch in s:
        st.push(ch)
    return st


def stack_tu_tu(s):
    """Mỗi phần tử Stack là một TỪ (cách nhau bởi khoảng trắng)."""
    st = StackList()
    for w in s.split():
        st.push(w)
    return st


# ---------- 3: QuickSort dùng Stack (khử đệ quy) ----------
def quick_sort_stack(arr):
    """QuickSort không đệ quy: dùng Stack lưu các cặp (lo, hi) cần sắp."""
    a = arr[:]
    if len(a) < 2:
        return a
    st = StackList()
    st.push((0, len(a) - 1))
    while not st.is_empty():
        lo, hi = st.pop()
        if lo >= hi:
            continue
        # phân hoạch Lomuto, pivot = a[hi]
        pivot = a[hi]
        i = lo - 1
        for j in range(lo, hi):
            if a[j] <= pivot:
                i += 1
                a[i], a[j] = a[j], a[i]
        i += 1
        a[i], a[hi] = a[hi], a[i]
        # đẩy 2 nửa vào stack
        st.push((lo, i - 1))
        st.push((i + 1, hi))
    return a


# ---------- 4: Infix -> Postfix (đã có ở BT5/BT7) ----------
print("1. Stack từ chuỗi ký tự 'Cau truc du lieu':")
st1 = stack_tu_ky_tu("Cau truc du lieu")
ra = []
while not st1.is_empty():
    ra.append(st1.pop())
print("   Pop ra:", "".join(ra), "(đảo của chuỗi gốc)\n")

print("2. Stack từ chuỗi 'Cau truc du lieu' (mỗi phần tử là 1 từ):")
st2 = stack_tu_tu("Cau truc du lieu")
ra = []
while not st2.is_empty():
    ra.append(st2.pop())
print("   Pop ra:", ra, "\n")

print("3. QuickSort dùng Stack:")
mang = [5, 2, 9, 1, 7, 3, 8, 4, 6, 0]
print("   Trước:", mang)
print("   Sau  :", quick_sort_stack(mang), "\n")

print("4. Infix -> Postfix (BT11):")
for bt in ["3+6*(2+3)", "10*2+(5*(8+4))-3", "(6+7)*2+((2+3)*(8+4))"]:
    print(f"   {bt:<28} -> {' '.join(infix_to_postfix(bt))}")
```

## BT12 (tr. 91) — Queue bằng mảng vòng

Cài đặt Queue sử dụng **mảng vòng** (circular array). Sinh ngẫu nhiên dãy n số, in ra và lần lượt đẩy vào Queue; đọc dãy số từ Queue và in ra.

**Mảng vòng** giải quyết nhược điểm "lãng phí phía trước" mà slide nêu ở tr. 86: khi `rear` chạm cuối mảng thì quay về 0 nếu còn chỗ.

```python
class QueueVong:
    """Hàng đợi cài bằng mảng vòng (FIFO)."""

    def __init__(self, capacity=8):
        self.cap = capacity
        self.a = [None] * capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == self.cap

    def enqueue(self, x):
        if self.is_full():
            return False
        self.a[self.rear] = x
        self.rear = (self.rear + 1) % self.cap      # quay vòng
        self.count += 1
        return True

    def dequeue(self):
        if self.is_empty():
            return None
        x = self.a[self.front]
        self.front = (self.front + 1) % self.cap    # quay vòng
        self.count -= 1
        return x

    def peek_front(self):
        return None if self.is_empty() else self.a[self.front]

    def peek_rear(self):
        return None if self.is_empty() else self.a[(self.rear - 1) % self.cap]

    def to_list(self):
        return [self.a[(self.front + i) % self.cap] for i in range(self.count)]

    def __len__(self):
        return self.count


import random
random.seed(2024)
n = 12
day = [random.randint(1, 99) for _ in range(n)]
print("Dãy sinh ngẫu nhiên:", day)

q = QueueVong(capacity=8)
print("\nĐẩy lần lượt vào Queue (capacity = 8):")
for x in day:
    if q.enqueue(x):
        print(f"   enqueue({x:>2}) -> Queue = {q.to_list()}")
    else:
        print(f"   enqueue({x:>2}) -> ĐẦY, bỏ qua (Queue = {q.to_list()})")

print("\nĐọc dãy số từ Queue:")
ra = []
while not q.is_empty():
    x = q.dequeue()
    ra.append(x)
    print(f"   dequeue() -> {x:>2} | còn lại {q.to_list()}")
print("\nDãy đọc ra:", ra)
```

```python
# ---- KIỂM THỬ BT12 ----
from collections import deque

# 1) Đối chiếu QueueVong với collections.deque qua thao tác ngẫu nhiên
random.seed(17)
q = QueueVong(capacity=7)
ref = deque()
ok = True
for _ in range(1000):
    if random.random() < 0.6:
        x = random.randint(0, 99)
        r1 = q.enqueue(x)
        r2 = len(ref) < 7
        if r2: ref.append(x)
        if r1 != r2:
            ok = False; print("enqueue lệch"); break
    else:
        r1 = q.dequeue()
        r2 = ref.popleft() if ref else None
        if r1 != r2:
            ok = False; print("dequeue lệch", r1, r2); break
    if q.to_list() != list(ref):
        ok = False; print("nội dung lệch"); break
print("Test QueueVong == deque (1000 thao tác, capacity=7):", "PASS" if ok else "FAIL")

# 2) Mảng vòng thực sự quay vòng: sau nhiều enqueue/dequeue, front và rear phải "wrap"
q = QueueVong(capacity=4)
for i in range(3): q.enqueue(i)
for _ in range(3): q.dequeue()
for i in range(3): q.enqueue(100 + i)
assert q.to_list() == [100, 101, 102]
assert q.front == 3 or q.rear == 2, f"front={q.front}, rear={q.rear} (phải đã quay vòng)"
print(f"Test quay vòng mảng (front={q.front}, rear={q.rear}): PASS")

# 3) FIFO: thứ tự ra = thứ tự vào
q = QueueVong(capacity=10)
day = [5, 3, 8, 1, 9]
for x in day: q.enqueue(x)
ra = [q.dequeue() for _ in range(len(day))]
assert ra == day
print("Test FIFO (thứ tự ra == thứ tự vào): PASS")

# 4) Biên: rỗng, đầy, peek
q = QueueVong(capacity=2)
assert q.is_empty() and q.dequeue() is None and q.peek_front() is None and q.peek_rear() is None
assert q.enqueue(1) and q.enqueue(2) and q.is_full() and q.enqueue(3) is False
assert q.peek_front() == 1 and q.peek_rear() == 2
print("Test biên Queue vòng (rỗng/đầy/peek front-rear): PASS")
```

## BT13 (tr. 92) — Các thao tác xử lý Queue

1. Nhập giá trị vào Queue
2. Lấy **5 giá trị đầu tiên** của Queue và xuất ra (nếu không đủ thì lấy đến hết)
3. Cho biết giá trị **đầu** và **cuối** Queue

```python
def bai_13(ds_input, capacity=20):
    q = QueueVong(capacity)
    # 1. Nhập giá trị vào Queue
    for x in ds_input:
        q.enqueue(x)
    print("1. Queue sau khi nhập:", q.to_list())

    # 3. Giá trị đầu và cuối
    print(f"3. Giá trị ĐẦU Queue: {q.peek_front()} | CUỐI Queue: {q.peek_rear()}")

    # 2. Lấy 5 giá trị đầu tiên (nếu không đủ thì lấy đến hết)
    lay = []
    for _ in range(5):
        if q.is_empty():
            break
        lay.append(q.dequeue())
    print(f"2. Lấy ra {len(lay)} giá trị đầu tiên: {lay}")
    print(f"   Queue còn lại: {q.to_list()}")
    return lay


print("=== Trường hợp Queue có NHIỀU hơn 5 phần tử ===")
bai_13([10, 20, 30, 40, 50, 60, 70, 80])

print("\n=== Trường hợp Queue có ÍT hơn 5 phần tử (3 phần tử) ===")
bai_13([1, 2, 3])

print("\n=== Trường hợp Queue RỖNG ===")
bai_13([])
```

```python
# ---- KIỂM THỬ BT13 ----
# 1) Lấy đúng 5 phần tử đầu khi Queue đủ lớn
lay = bai_13([10, 20, 30, 40, 50, 60, 70, 80])
assert lay == [10, 20, 30, 40, 50]
print("Test lấy đúng 5 giá trị đầu (FIFO): PASS")

# 2) Queue ít hơn 5 -> lấy đến hết
lay = bai_13([1, 2, 3])
assert lay == [1, 2, 3]
print("Test Queue < 5 phần tử -> lấy đến hết: PASS")

# 3) Queue rỗng -> lấy ra rỗng, không lỗi
lay = bai_13([])
assert lay == []
print("Test Queue rỗng -> lấy ra []: PASS")

# 4) Đối chiếu độc lập: giá trị đầu/cuối
q = QueueVong(20)
for x in [7, 3, 9, 1]: q.enqueue(x)
assert q.peek_front() == 7 and q.peek_rear() == 1
print("Test giá trị đầu = 7, cuối = 1: PASS")

# 5) Sau khi lấy 5, phần còn lại đúng thứ tự
q = QueueVong(20)
for x in range(1, 9): q.enqueue(x)      # 1..8
lay = [q.dequeue() for _ in range(5)]
assert lay == [1, 2, 3, 4, 5] and q.to_list() == [6, 7, 8]
print("Test phần còn lại sau khi lấy 5 == [6,7,8]: PASS")
```

## BT14 (tr. 93) — Mô phỏng hàng chờ ngân hàng bằng Queue (DSLK đơn)

Mô phỏng hàng chờ khách hàng tại ngân hàng, FIFO. Mỗi khách có **mã ID** và **loại giao dịch** (rút tiền, gửi tiền, kiểm tra số dư). Ngân hàng có **một quầy**.

Các thao tác cần có:
- `addCustomer(customerId, transactionType)` — thêm khách vào hàng chờ
- `serveCustomer()` — phục vụ khách đầu tiên rồi xóa khỏi hàng
- `nextCustomer()` — xem khách tiếp theo **không** xóa
- `isQueueEmpty()` — kiểm tra hàng chờ trống

```python
class KhachHang:
    def __init__(self, ma, loai_gd):
        self.ma = ma
        self.loai_gd = loai_gd

    def __str__(self):
        return f"KH{self.ma}({self.loai_gd})"


class QueueDSLK:
    """Hàng đợi cài bằng danh sách liên kết đơn (front = head, rear = tail)."""

    def __init__(self):
        self.front = None
        self.rear = None
        self.n = 0

    def is_empty(self):
        return self.front is None

    def enqueue(self, x):
        p = Node(x)
        if self.rear is None:
            self.front = self.rear = p
        else:
            self.rear.next = p
            self.rear = p
        self.n += 1

    def dequeue(self):
        if self.front is None:
            return None
        x = self.front.info
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        self.n -= 1
        return x

    def peek(self):
        return None if self.front is None else self.front.info

    def to_list(self):
        out, p = [], self.front
        while p:
            out.append(p.info)
            p = p.next
        return out

    def __len__(self):
        return self.n


class NganHang:
    """Mô phỏng hàng chờ ngân hàng với 1 quầy phục vụ (BT14)."""

    def __init__(self):
        self.hang_cho = QueueDSLK()
        self.lich_su = []

    # addCustomer
    def addCustomer(self, customerId, transactionType):
        self.hang_cho.enqueue(KhachHang(customerId, transactionType))
        print(f"  + Thêm {KhachHang(customerId, transactionType)} vào hàng chờ "
              f"(hàng chờ: {len(self.hang_cho)} khách)")

    # serveCustomer
    def serveCustomer(self):
        if self.hang_cho.is_empty():
            print("  ! Hàng chờ trống, không có ai để phục vụ.")
            return None
        kh = self.hang_cho.dequeue()
        self.lich_su.append(kh)
        print(f"  > Đang phục vụ {kh} ... xong. Còn lại {len(self.hang_cho)} khách.")
        return kh

    # nextCustomer
    def nextCustomer(self):
        kh = self.hang_cho.peek()
        print(f"  ? Khách tiếp theo: {kh if kh else '(không có)'}")
        return kh

    # isQueueEmpty
    def isQueueEmpty(self):
        r = self.hang_cho.is_empty()
        print(f"  ? Hàng chờ trống? {r}")
        return r


print("=== MÔ PHỎNG NGÂN HÀNG ===")
nh = NganHang()
nh.isQueueEmpty()
nh.addCustomer(101, "rút tiền")
nh.addCustomer(102, "gửi tiền")
nh.addCustomer(103, "kiểm tra số dư")
nh.addCustomer(104, "rút tiền")
print()
nh.nextCustomer()
nh.isQueueEmpty()
print()
nh.serveCustomer()
nh.nextCustomer()
nh.serveCustomer()
nh.serveCustomer()
nh.serveCustomer()      # hàng đã hết
nh.isQueueEmpty()
print("\nLịch sử phục vụ:", [str(k) for k in nh.lich_su])
```

```python
# ---- KIỂM THỬ BT14 ----
nh = NganHang()

# 1) Hàng chờ trống ban đầu
assert nh.isQueueEmpty() is True
print("Test hàng chờ trống ban đầu: PASS")

# 2) Thêm khách -> FIFO
nh.addCustomer(1, "rút tiền")
nh.addCustomer(2, "gửi tiền")
nh.addCustomer(3, "kiểm tra số dư")
assert nh.isQueueEmpty() is False
assert len(nh.hang_cho) == 3
print("Test thêm 3 khách: PASS")

# 3) nextCustomer KHÔNG xóa khỏi hàng
kh = nh.nextCustomer()
assert kh.ma == 1 and len(nh.hang_cho) == 3, "nextCustomer không được xóa khách"
print("Test nextCustomer không xóa khách khỏi hàng: PASS")

# 4) serveCustomer phục vụ đúng thứ tự FIFO
k1 = nh.serveCustomer(); assert k1.ma == 1
k2 = nh.serveCustomer(); assert k2.ma == 2
k3 = nh.serveCustomer(); assert k3.ma == 3
assert nh.isQueueEmpty() is True
assert [k.ma for k in nh.lich_su] == [1, 2, 3]
print("Test serveCustomer theo đúng thứ tự FIFO 1, 2, 3: PASS")

# 5) Phục vụ khi hàng trống -> trả None, không lỗi
assert nh.serveCustomer() is None
print("Test serveCustomer khi hàng trống (trả None): PASS")

# 6) Đối chiếu với collections.deque
from collections import deque
nh2 = NganHang(); ref = deque()
random.seed(23)
for _ in range(200):
    if random.random() < 0.6:
        ma = random.randint(1, 999)
        nh2.addCustomer(ma, "rút tiền"); ref.append(ma)
    else:
        k = nh2.serveCustomer()
        r = ref.popleft() if ref else None
        assert (k.ma if k else None) == r
    assert [x.ma for x in nh2.hang_cho.to_list()] == list(ref)
print("Test NganHang == deque (200 thao tác ngẫu nhiên): PASS")

# 7) Queue DSLK: FIFO và biên
q = QueueDSLK()
assert q.is_empty() and q.dequeue() is None and q.peek() is None
for x in [1, 2, 3]: q.enqueue(x)
assert q.to_list() == [1, 2, 3] and q.peek() == 1
assert [q.dequeue() for _ in range(3)] == [1, 2, 3]
assert q.is_empty() and q.rear is None
print("Test QueueDSLK (FIFO, rỗng, rear reset sau khi hết): PASS")
```

## BT15 (tr. 95) — Các thao tác cơ bản của hàng đợi số nguyên dùng DSLK đơn

1. Khởi tạo: `Init(Queue &q)`
2. Kiểm tra rỗng: `IsEmpty(Queue q)`
3. Thêm phần tử: `EnQueue(Queue &q, int newItem)`
4. Lấy phần tử: `DeQueue(Queue &q, int &outItem)`
5. Xem phần tử đầu: `Front(Queue q, int &outItem)`
6. Xem phần tử cuối: `Rear(Queue q, int &outItem)`

```python
class QueueSoNguyen:
    """Hàng đợi số nguyên dùng DSLK đơn — đủ 6 thao tác của BT15."""

    def __init__(self):
        self.front = None
        self.rear = None
        self.n = 0

    # 1. Init
    def Init(self):
        self.front = self.rear = None
        self.n = 0

    # 2. IsEmpty
    def IsEmpty(self):
        return self.front is None

    # 3. EnQueue
    def EnQueue(self, newItem):
        p = Node(newItem)
        if self.rear is None:
            self.front = self.rear = p
        else:
            self.rear.next = p
            self.rear = p
        self.n += 1
        return True

    # 4. DeQueue
    def DeQueue(self):
        """Trả về (True, outItem) nếu lấy được, (False, None) nếu hàng rỗng."""
        if self.front is None:
            return False, None
        out = self.front.info
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        self.n -= 1
        return True, out

    # 5. Front
    def Front(self):
        if self.front is None:
            return False, None
        return True, self.front.info

    # 6. Rear
    def Rear(self):
        if self.rear is None:
            return False, None
        return True, self.rear.info

    def to_list(self):
        out, p = [], self.front
        while p:
            out.append(p.info)
            p = p.next
        return out


q = QueueSoNguyen()
print("1. Init -> IsEmpty =", q.IsEmpty())
print("\n3. EnQueue lần lượt 5, 8, 3, 9:")
for x in [5, 8, 3, 9]:
    print(f"   EnQueue({x}) -> {q.EnQueue(x)} | Queue = {q.to_list()}")

print("\n2. IsEmpty =", q.IsEmpty())
print("5. Front    =", q.Front())
print("6. Rear     =", q.Rear())

print("\n4. DeQueue:")
while True:
    ok, x = q.DeQueue()
    if not ok:
        print("   Hàng rỗng -> DeQueue trả False")
        break
    print(f"   DeQueue() -> ({ok}, {x}) | còn lại {q.to_list()}")

print("\nSau khi hết: IsEmpty =", q.IsEmpty(), "| Front =", q.Front(), "| Rear =", q.Rear())
```

```python
# ---- KIỂM THỬ BT15 ----
from collections import deque

# 1) Init
q = QueueSoNguyen()
q.EnQueue(1); q.EnQueue(2)
q.Init()
assert q.IsEmpty() and q.to_list() == [] and q.n == 0
print("Test Init xóa sạch hàng đợi: PASS")

# 2) IsEmpty
assert q.IsEmpty() is True
q.EnQueue(5)
assert q.IsEmpty() is False
print("Test IsEmpty: PASS")

# 3) EnQueue luôn trả True, thêm vào CUỐI
q = QueueSoNguyen()
for x in [1, 2, 3]: assert q.EnQueue(x) is True
assert q.to_list() == [1, 2, 3]
print("Test EnQueue (thêm vào cuối): PASS")

# 4) DeQueue lấy từ ĐẦU, trả (True, x); rỗng -> (False, None)
ok, x = q.DeQueue()
assert ok is True and x == 1 and q.to_list() == [2, 3]
ok, x = q.DeQueue(); ok2, x2 = q.DeQueue()
assert ok and ok2
ok, x = q.DeQueue()
assert ok is False and x is None
print("Test DeQueue (lấy từ đầu, rỗng trả False): PASS")

# 5) Front / Rear
q = QueueSoNguyen()
ok, x = q.Front(); assert ok is False and x is None
ok, x = q.Rear();  assert ok is False and x is None
for v in [7, 3, 9]: q.EnQueue(v)
ok, x = q.Front(); assert ok and x == 7
ok, x = q.Rear();  assert ok and x == 9
print("Test Front/Rear (rỗng -> False; có dữ liệu -> 7 và 9): PASS")

# 6) Đối chiếu toàn diện với deque
random.seed(29)
q = QueueSoNguyen(); ref = deque()
ok_all = True
for _ in range(800):
    r = random.random()
    if r < 0.55:
        x = random.randint(0, 99)
        assert q.EnQueue(x) is True
        ref.append(x)
    elif r < 0.9:
        o1, v1 = q.DeQueue()
        o2 = len(ref) > 0
        v2 = ref.popleft() if o2 else None
        if o1 != o2 or v1 != v2:
            ok_all = False; print("DeQueue lệch"); break
    else:
        o1, v1 = q.Front(); o2, v2 = q.Rear()
        exp_f = ref[0] if ref else None
        exp_r = ref[-1] if ref else None
        if (o1, v1, o2, v2) != (bool(ref), exp_f, bool(ref), exp_r):
            ok_all = False; print("Front/Rear lệch"); break
    if q.to_list() != list(ref):
        ok_all = False; print("nội dung lệch"); break
print("Test QueueSoNguyen == deque (800 thao tác, đủ 6 hàm):", "PASS" if ok_all else "FAIL")

# 7) rear phải được reset khi hàng trở về rỗng
q = QueueSoNguyen()
q.EnQueue(1); q.DeQueue()
assert q.rear is None and q.front is None
print("Test rear/front reset khi hàng rỗng: PASS")
```

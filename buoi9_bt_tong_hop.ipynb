## Bài tập 1: Biểu diễn đồ thị bằng danh sách kề

### 1. Đề bài
Viết chương trình Python để:
1. Nhập số đỉnh và danh sách cạnh.
2. Biểu diễn đồ thị bằng danh sách kề.
3. In danh sách kề của từng đỉnh.

### 2. Ý tưởng & Cấu trúc dữ liệu
- **Danh sách kề (Adjacency List):** Sử dụng `dict` (hoặc `list`) trong Python để lưu trữ danh sách các đỉnh kề với mỗi đỉnh `u`.
- Với đồ thị vô hướng, mỗi cạnh `(u, v)` sẽ được thêm 2 chiều:
  - Thêm `v` vào danh sách kề của `u`.
  - Thêm `u` vào danh sách kề của `v`.
- **Độ phức tạp:**
  - Bộ nhớ: $O(V + E)$
  - Thời gian khởi tạo: $O(V + E)$


```python
def create_adjacency_list(num_vertices, edges, is_directed=False):
    """
    Hàm tạo và trả về danh sách kề của đồ thị.
    - num_vertices: Số lượng đỉnh của đồ thị
    - edges: Danh sách các cạnh dạng tuple (u, v)
    - is_directed: True nếu là đồ thị có hướng, False nếu là vô hướng
    """
    adj_list = {i: [] for i in range(num_vertices)}
    for u, v in edges:
        adj_list[u].append(v)
        if not is_directed:
            adj_list[v].append(u)
            
    # Sắp xếp các đỉnh kề để in ra thứ tự tăng dần
    for u in adj_list:
        adj_list[u].sort()
    return adj_list

def print_adjacency_list(adj_list):
    """In danh sách kề của từng đỉnh."""
    for u in sorted(adj_list.keys()):
        print(f"{u}: {adj_list[u]}")

# --- Test case ---
num_vertices = 4
edges = [(0, 1), (0, 2), (1, 2), (2, 3)]

print("Danh sách kề của đồ thị:")
adj_list = create_adjacency_list(num_vertices, edges)
print_adjacency_list(adj_list)
```

## Bài tập 2: Duyệt đồ thị theo chiều rộng (BFS - Breadth-First Search)

### 1. Đề bài
1. Sử dụng danh sách kề để biểu diễn đồ thị.
2. Viết hàm `bfs(start)` duyệt theo chiều rộng từ đỉnh `start`.

### 2. Ý tưởng & Thuật toán
- **BFS (Breadth-First Search):** Duyệt đồ thị theo từng mức/lớp láng giềng tỏa ra từ đỉnh nguồn `start`.
- Sử dụng cấu trúc hàng đợi **Queue (FIFO - First In First Out)** thông qua `collections.deque` và một tập hợp `visited` để đánh dấu các đỉnh đã thăm.
- **Các bước thực hiện:**
  1. Đẩy đỉnh `start` vào hàng đợi và thêm vào `visited`.
  2. Lặp lại khi hàng đợi còn phần tử:
     - Lấy đỉnh `u` ở đầu hàng đợi ra (popleft) và ghi nhận thứ tự duyệt.
     - Với mỗi đỉnh kề `v` của `u`, nếu `v` chưa thăm: đánh dấu `visited` và đưa vào hàng đợi.
- **Độ phức tạp:** $O(V + E)$


```python
from collections import deque

def build_graph(edges, is_directed=False):
    """Xây dựng danh sách kề từ danh sách cạnh."""
    adj = {}
    for u, v in edges:
        adj.setdefault(u, []).append(v)
        if not is_directed:
            adj.setdefault(v, []).append(u)
    for u in adj:
        adj[u].sort()
    return adj

def bfs(graph, start):
    """
    Duyệt đồ thị theo chiều rộng (BFS) từ đỉnh start.
    Trả về danh sách thứ tự các đỉnh được duyệt.
    """
    visited = {start}
    queue = deque([start])
    traversal_order = []
    
    while queue:
        u = queue.popleft()
        traversal_order.append(u)
        
        for v in graph.get(u, []):
            if v not in visited:
                visited.add(v)
                queue.append(v)
                
    return traversal_order

# --- Test case ---
edges_bfs = [(0, 1), (0, 2), (1, 3), (2, 3), (3, 4)]
graph_bfs = build_graph(edges_bfs, is_directed=False)

start_vertex = 0
result_bfs = bfs(graph_bfs, start_vertex)

print("Thứ tự duyệt BFS từ đỉnh", start_vertex, ":")
print(" ".join(map(str, result_bfs)))
```

## Bài tập 3: Duyệt đồ thị theo chiều sâu (DFS - Depth-First Search)

### 1. Đề bài
1. Viết hàm `dfs(start)` duyệt đồ thị theo chiều sâu (dùng đệ quy).
2. In thứ tự các đỉnh được duyệt.

### 2. Ý tưởng & Thuật toán
- **DFS (Depth-First Search):** Duyệt theo một nhánh sâu nhất có thể trước khi quay lui (Backtracking).
- Cài đặt bằng **đệ quy**, hệ thống sẽ tự động sử dụng **Call Stack** để quản lý quá trình duyệt.
- **Các bước thực hiện:**
  1. Khi gọi `dfs(u)`, đánh dấu `u` đã thăm và lưu vào danh sách thứ tự duyệt.
  2. Với mỗi đỉnh kề `v` của `u`: nếu `v` chưa được thăm, gọi đệ quy `dfs(v)`.
- **Độ phức tạp:** $O(V + E)$


```python
def build_graph_dfs(edges, is_directed=False):
    """Tạo danh sách kề cho đồ thị."""
    adj = {}
    for u, v in edges:
        adj.setdefault(u, []).append(v)
        if not is_directed:
            adj.setdefault(v, []).append(u)
    for u in adj:
        adj[u].sort()
    return adj

def dfs_recursive(graph, start):
    """
    Duyệt đồ thị theo chiều sâu (DFS) sử dụng đệ quy.
    Trả về danh sách thứ tự các đỉnh được duyệt.
    """
    visited = set()
    traversal_order = []
    
    def dfs(u):
        visited.add(u)
        traversal_order.append(u)
        for v in graph.get(u, []):
            if v not in visited:
                dfs(v)
                
    dfs(start)
    return traversal_order

# --- Test case ---
edges_dfs = [(1, 2), (1, 3), (2, 4), (3, 5), (5, 6)]
graph_dfs = build_graph_dfs(edges_dfs, is_directed=False)

start_dfs = 1
result_dfs = dfs_recursive(graph_dfs, start_dfs)

print("Thứ tự duyệt DFS từ đỉnh", start_dfs, ":")
print(" ".join(map(str, result_dfs)))
```

## Bài tập 4: Tìm đường đi ngắn nhất bằng thuật toán Dijkstra

### 1. Đề bài
1. Biểu diễn đồ thị có trọng số bằng ma trận kề hoặc danh sách kề.
2. Viết hàm `dijkstra(graph, start)` để tính khoảng cách ngắn nhất từ `start` đến tất cả các đỉnh.

### 2. Ý tưởng & Thuật toán
- **Thuật toán Dijkstra:** Tìm đường đi ngắn nhất từ một đỉnh nguồn đến tất cả các đỉnh khác trong đồ thị có trọng số không âm.
- Sử dụng cấu trúc **Hàng đợi ưu tiên (Priority Queue / Min-Heap)** thông qua thư viện `heapq` trong Python để luôn trích xuất đỉnh có khoảng cách tạm thời nhỏ nhất.
- **Nguyên lý cập nhật khoảng cách (Relaxation):**
  - Khởi tạo khoảng cách ban đầu: $dist[start] = 0$, các đỉnh còn lại là $\infty$.
  - Nếu $dist[u] + weight(u, v) < dist[v]$:
    - Cập nhật $dist[v] = dist[u] + weight(u, v)$
    - Đẩy $(dist[v], v)$ vào hàng đợi ưu tiên.
- **Độ phức tạp:** $O((V + E) \log V)$


```python
import heapq

def dijkstra(graph, start, num_vertices):
    """
    Tính khoảng cách ngắn nhất từ đỉnh start đến tất cả các đỉnh bằng Dijkstra.
    - graph: dictionary dạng {u: [(v, weight), ...]}
    - start: đỉnh bắt đầu
    - num_vertices: số lượng đỉnh
    """
    # Khởi tạo khoảng cách ban đầu là vô cực
    distances = [float('inf')] * num_vertices
    distances[start] = 0
    
    # Priority Queue lưu tuple: (khoảng_cách, đỉnh)
    pq = [(0, start)]
    
    while pq:
        curr_dist, u = heapq.heappop(pq)
        
        # Bỏ qua nếu khoảng cách này lớn hơn khoảng cách ngắn nhất đã tìm thấy
        if curr_dist > distances[u]:
            continue
            
        for v, weight in graph.get(u, []):
            new_dist = curr_dist + weight
            if new_dist < distances[v]:
                distances[v] = new_dist
                heapq.heappush(pq, (new_dist, v))
                
    return distances

# --- Test case ---
# Đồ thị có hướng có trọng số:
# 0 -1-> 1, 0 -4-> 2, 1 -2-> 2, 1 -6-> 3, 2 -3-> 3
num_vertices_dijkstra = 4
weighted_edges = [
    (0, 1, 1),
    (0, 2, 4),
    (1, 2, 2),
    (1, 3, 6),
    (2, 3, 3)
]

# Biểu diễn bằng danh sách kề
graph_dijkstra = {i: [] for i in range(num_vertices_dijkstra)}
for u, v, w in weighted_edges:
    graph_dijkstra[u].append((v, w))

start_dijkstra = 0
distances = dijkstra(graph_dijkstra, start_dijkstra, num_vertices_dijkstra)

print(f"Khoảng cách từ {start_dijkstra} đến các đỉnh:")
print(distances)
```

## Bài tập 5: Tìm cây khung nhỏ nhất (MST) bằng thuật toán Kruskal

### 1. Đề bài
1. Cài đặt thuật toán Kruskal để tìm cây khung nhỏ nhất (Minimum Spanning Tree - MST).
2. In các cạnh được chọn và tổng trọng số.

### 2. Ý tưởng & Thuật toán
- **Cây khung nhỏ nhất (MST):** Một tập hợp gồm $V - 1$ cạnh nối tất cả $V$ đỉnh với nhau sao cho tổng trọng số là nhỏ nhất và không tạo thành chu trình.
- **Thuật toán Kruskal (Tham lam - Greedy):**
  1. Sắp xếp toàn bộ các cạnh theo thứ tự trọng số tăng dần.
  2. Lần lượt xét từng cạnh: nếu 2 đầu mút của cạnh thuộc 2 tập hợp khác nhau (không tạo thành chu trình), ta chọn cạnh đó vào MST và hợp nhất 2 tập hợp.
  3. Quá trình dừng khi đã chọn đủ $V - 1$ cạnh hoặc đã duyệt hết các cạnh.
- **Cấu trúc dữ liệu Disjoint Set Union (DSU / Union-Find):**
  - Thao tác `find(i)`: Tìm gốc của phần tử `i` kết hợp **Path Compression** (nén đường đi).
  - Thao tác `union(i, j)`: Hợp nhất hai tập hợp chứa `i` và `j`. Trả về `False` nếu chúng đã cùng thuộc một tập hợp (phát hiện chu trình).
- **Độ phức tạp:** $O(E \log E)$


```python
class DSU:
    """Cấu trúc dữ liệu Disjoint Set Union (Union-Find) với Path Compression."""
    def __init__(self, n):
        self.parent = list(range(n))
        
    def find(self, i):
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i])  # Path compression
        return self.parent[i]
        
    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)
        if root_i != root_j:
            self.parent[root_i] = root_j
            return True
        return False  # Cùng gốc -> thêm cạnh này sẽ tạo chu trình

def kruskal(num_vertices, edges):
    """
    Tìm cây khung nhỏ nhất bằng thuật toán Kruskal.
    - num_vertices: Số lượng đỉnh của đồ thị
    - edges: Danh sách các cạnh dạng tuple (u, v, weight)
    """
    # Bước 1: Sắp xếp các cạnh theo trọng số tăng dần
    sorted_edges = sorted(edges, key=lambda edge: edge[2])
    
    dsu = DSU(num_vertices)
    mst_edges = []
    total_weight = 0
    
    for u, v, weight in sorted_edges:
        # Bước 2: Kiểm tra nếu không tạo chu trình thì thêm vào MST
        if dsu.union(u, v):
            mst_edges.append((u, v))
            total_weight += weight
            # Dừng sớm nếu đã chọn đủ V - 1 cạnh
            if len(mst_edges) == num_vertices - 1:
                break
                
    return mst_edges, total_weight

# --- Test case ---
num_vertices_kruskal = 4
edges_kruskal = [
    (0, 1, 10),
    (0, 2, 6),
    (0, 3, 5),
    (1, 3, 15),
    (2, 3, 4)
]

mst, total_weight = kruskal(num_vertices_kruskal, edges_kruskal)

print("Các cạnh trong MST:", ", ".join(str(e) for e in mst))
print("Tổng trọng số:", total_weight)
```

### Test kinh điển 2 (Bài 2 & 3): So sánh BFS (Theo tầng) vs DFS (Theo nhánh) trên Cây phân cấp
- **Cấu trúc cây nhị phân:**
  - Gốc `1` nối với `2` và `3`
  - Đỉnh `2` nối với `4` và `5`
  - Đỉnh `3` nối với `6` và `7`
- **Kỳ vọng:**
  - **BFS (Level-Order):** Duyệt tuần tự từng tầng từ trên xuống dưới $\rightarrow$ `1 2 3 4 5 6 7`.
  - **DFS (Pre-Order):** Đi sâu hết nhánh bên trái rồi mới quay lui sang nhánh phải $\rightarrow$ `1 2 4 5 3 6 7`.


```python
# Cây nhị phân gồm 7 đỉnh
tree_edges = [
    (1, 2), (1, 3),
    (2, 4), (2, 5),
    (3, 6), (3, 7)
]
tree_graph = build_graph(tree_edges, is_directed=False)

bfs_order = bfs(tree_graph, start=1)
dfs_order = dfs_recursive(tree_graph, start=1)

print("Cây phân cấp 7 nút (gốc 1):")
print("- Thứ tự duyệt BFS (theo tầng) :", " ".join(map(str, bfs_order)))
print("- Thứ tự duyệt DFS (theo nhánh):", " ".join(map(str, dfs_order)))
```

### Test kinh điển 3 (Bài 4): Đồ thị 9 đỉnh kinh điển (Sách giáo khoa CLRS - Introduction to Algorithms)
- **Đặc điểm:** Đồ thị có 9 đỉnh (0 đến 8) với 14 cạnh có trọng số phức tạp.
- **Mục tiêu kiểm thử:** Thuật toán Dijkstra phải chọn chính xác đường đi tối ưu đến mọi đỉnh, xử lý đúng các trường hợp đường đi vòng nhiều cạnh nhưng tổng trọng số lại nhỏ hơn đường đi ít cạnh.


```python
# Test case đồ thị 9 đỉnh kinh điển (CLRS)
clrs_num_vertices = 9
clrs_edges = [
    (0, 1, 4), (0, 7, 8),
    (1, 2, 8), (1, 7, 11),
    (2, 3, 7), (2, 8, 2), (2, 5, 4),
    (3, 4, 9), (3, 5, 14),
    (4, 5, 10),
    (5, 6, 2),
    (6, 7, 1), (6, 8, 6),
    (7, 8, 7)
]

# Đồ thị vô hướng có trọng số
clrs_graph = {i: [] for i in range(clrs_num_vertices)}
for u, v, w in clrs_edges:
    clrs_graph[u].append((v, w))
    clrs_graph[v].append((u, w))

clrs_distances = dijkstra(clrs_graph, start=0, num_vertices=clrs_num_vertices)

print("Khoảng cách ngắn nhất từ đỉnh 0 đến các đỉnh (0 -> 8):")
for node, d in enumerate(clrs_distances):
    print(f"  Đỉnh {node}: {d}")

print("\n-> Mảng kết quả chuẩn:", clrs_distances)
```

### Test kinh điển 4 (Bài 4): Đồ thị có đỉnh không liên thông (Unreachable Node)
- **Đặc điểm:** Đỉnh 3 không có cạnh nào dẫn tới từ các đỉnh khác.
- **Kỳ vọng:** Khoảng cách đến đỉnh 3 phải là vô cực (`inf`).


```python
# Đồ thị có hướng: 0 -> 1 -> 2, đỉnh 3 bị cô lập
unreachable_edges = [(0, 1, 5), (1, 2, 3)]
unreach_graph = {i: [] for i in range(4)}
for u, v, w in unreachable_edges:
    unreach_graph[u].append((v, w))

dist_unreach = dijkstra(unreach_graph, start=0, num_vertices=4)
print("Khoảng cách từ 0 khi đỉnh 3 không liên thông:")
print("Kết quả:", dist_unreach)
```

### Test kinh điển 5 (Bài 5): Cây khung nhỏ nhất (MST) trên đồ thị CLRS (9 đỉnh, 14 cạnh)
- **Đặc điểm:** Thuật toán Kruskal phải chọn ra đúng $V - 1 = 8$ cạnh sao cho tổng trọng số là nhỏ nhất và tự động bỏ qua 6 cạnh còn lại vì chúng tạo thành chu trình.
- **Kỳ vọng chuẩn từ sách CLRS:** Tổng trọng số nhỏ nhất là **37**.


```python
# Chạy thuật toán Kruskal trên đồ thị CLRS 9 đỉnh
mst_clrs, total_w_clrs = kruskal(clrs_num_vertices, clrs_edges)

print(f"Tổng số đỉnh: {clrs_num_vertices} -> Số cạnh trong MST phải bằng {clrs_num_vertices - 1}")
print(f"Số cạnh được chọn: {len(mst_clrs)}")
print("Danh sách các cạnh MST được chọn (theo thứ tự trọng số tăng dần):")
for u, v in mst_clrs:
    # Tìm lại trọng số của cạnh
    w = next(weight for (x, y, weight) in clrs_edges if (x, y) == (u, v) or (y, x) == (u, v))
    print(f"  Cạnh ({u}, {v}) - Trọng số: {w}")

print(f"\nTổng trọng số nhỏ nhất của cây khung (MST): {total_w_clrs}")
assert total_w_clrs == 37, "Tổng trọng số phải bằng 37!"
print("=> KẾT QUẢ ĐẠT CHUẨN 100%!")
```
